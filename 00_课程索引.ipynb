# Python 编程基础（智慧交通）· Jupyter 课件

> **适用对象**：智能网联汽车技术应用、汽车类、城轨类各专业（初中起点五年制高技）
>
> **课程结构**：13 章 + 1 个工具专题节（第 5A 节 · Trae 调试）+ 1 次终结性考核。每章统一按 **「知识讲解 → 代码示范 → 布置练习 → 练习答案」** 四段式组织，可打开即按顺序讲。
>
> **打开方式**：双击本目录下的 **`启动课件.command`**；或在本目录执行 `jupyter lab 00_课程索引.ipynb`。

## 一、每章怎么讲（四段式）

| 顺序 | 板块 | 内容构成 | 课堂用途 |
|---|---|---|---|
| 1 | **第一部分　知识讲解** | 知识点 1 ~ n，每个知识点含「核心概念 / 语法要点 / 常见易错点 / 随堂示例」 | 讲授，配合板书与提问 |
| 2 | **第二部分　代码示范** | 案例 ×2：任务情境 → 任务要求 → 参考代码（带注释）→ 运行结果 → 代码解读表 | 教师演示，学生跟做 |
| 3 | **第三部分　布置练习** | 练习 ×2：题目 → 要求 → 提示 → 自检要点（**不含答案**） | 课堂练习 / 课后作业 |
| 4 | **第四部分　习题答案** | 答案 ×2：参考代码 → 运行结果 → 解析 → 常见错误 → 评分要点 | 讲评，先做后看 |

**随堂示例**：全课程 74 个知识点，每个知识点后紧跟一个「核心要点 → 可运行代码（带中文注释）→ 预期输出」的小例子，共 **74 个**。代码单元格可直接 `Shift + Enter` 运行，文档中的预期输出就是它的真实 stdout，用于当堂验证结论。其中第 5A 节的 7 个示例以独立 `.py` 文件呈现——调试必须在 IDE 里设断点，所以它们既进讲义、也是 Trae 中的操作对象。

每章开头另有 **「本章学习路径」**（四段式索引表 + 知识点与代码对照表）与 **「教学提示（教师用）」**（建议课时、讲解主线、学生卡点、课堂活动）；章末有 **「本章小结」**「自检清单」「前后章节衔接」。

## 二、课时与课型建议

| 课型 | 时长 | 建议安排 |
|---|---|---|
| 标准 2 学时课 | 90 min | 讲解 25 min → 案例演示 25 min → 学生练习 30 min → 讲评 10 min |
| 8 学时章（第 3、5、6、8、11 章） | 4 次课 | 每次 2—3 个知识点 + 1 个案例 + 1 个练习，练习 2 留作课后作业 |
| 第 13 章项目周 | 12 学时 | 先 turtle 后 pygame；分组改造参数、录屏汇报 |
| 第 5A 节调试专题 | 4 学时 | 2 次课，全在机房/自带笔记本上操作：第 1 次课练断点与单步，第 2 次课完成两个带预设 bug 的案例 |

> 13 章合计 80 学时，与课程标准模块学时一一对应（见 `00课程说明与索引.md` 第七节）；第 5A 节 4 学时作为第 5、6 章之间的工具训练插入，全课程共 84 学时。

## 四、运行说明

| 场景 | 说明 |
|---|---|
| 运行一个单元格 | 选中后按 **Shift + Enter** |
| 运行全章 | 菜单 **Run → Run All Cells** |
| 结果异常 / 变量混乱 | **Kernel → Restart Kernel**，再重新运行 |
| 含 `input()` 的章节 | 第 2、3、4、9 章：运行后单元格下方出现输入框，按「运行结果（示例输入：…）」中给出的值填写 |
| 第 10 章多文件 | 标有「模块文件」的代码块在 Notebook 中带 `%%writefile`，运行即写盘；随后运行调用程序才能 `import` 成功 |
| 第 13 章图形界面 | `turtle` 为标准库；`pygame` 需 `pip install pygame`；绘图窗口需**手动关闭**才能继续，否则内核被占用 |
| 文件操作章节 | 第 8 章会在当前目录生成 `vehicle_log.txt`、`flow_daily.csv` 等，重复运行会覆盖，属正常现象 |
| 第 5A 节调试专题 | 断点、单步、变量与堆栈等操作**在 Trae 中完成**（打开 `05A_Trae调试功能` 文件夹，按 F5 启动调试）；Notebook 用于投屏讲解与对照「错误输出 / 正确输出」，其中代码单元格可直接运行看结果 |

## 五、两条使用纪律

1. **不超纲**：每章代码只使用该章及之前学过的语法。例如第 9 章不用自定义异常类（`class` 在第 11 章），第 7 章的装饰器用调用计数而非计时（`time` 模块在第 10 章）；第 5A 节只用第 1—5 章的语法，其中出现的 `def` 仅用于演示「步入 / 步出」，函数语法的系统讲解在第 6 章。
2. **先做后看**：第三部分的练习只给题目与提示，答案在第四部分。建议学生先跑通自己的版本、对照「自检要点」自查，再看答案的「解析」与「常见错误」。

> 各章「运行结果」为参考输出，实际以运行结果为准；中英文混排时 `\t` 与定宽对齐会出现轻微参差，属正常现象。
