# 第一部分　知识讲解

## 2.1 场景导入：交通指标都是"数值"

智慧交通的所有指标都是数值：车流密度、平均车速、通行能力、配时时长、通行费。程序必须先能**正确存放**这些数值（选对类型），再能**正确计算**它们（选对运算符）。

类型选错的后果很具体：`"120" + 5` 直接报 `TypeError`；`"120" * 5` 不报错却得到 `"120120120120120"`。这类"不报错但结果错"的问题比报错更危险，本章的一半篇幅都在解决它。

## 2.2 知识点 2.1　变量、赋值与命名

### 2.2.1 核心概念

**变量**是给数据起的名字。赋值语句 `speed = 62` 的含义是"把 62 存进名为 speed 的盒子"，此后写 `speed` 就等于写 62。

三条基本性质：

1. **变量不需要声明类型**：赋值即创建，类型由所赋的值决定（动态类型）；
2. **变量可以重新赋值**，新值会覆盖旧值，类型也可以改变（`x = 1` 之后 `x = "1"` 合法但不建议）；
3. **变量必须先赋值后使用**，否则 `NameError`。

**原理补充（概念深化）**：Python 的变量更接近「贴在数据上的标签」，而不是「装数据的盒子」。执行 `flow = 820` 时，解释器先在内存里创建一个整数对象，再把名字 `flow` 绑定到它；重新赋值 `flow = flow + 180` 并不是往原盒子里改数，而是**先读旧值算出新对象，再把名字改绑过去**。整数对象本身不可变，所以同一份 `820` 若被多个名字引用，谁都不会被别人改动。这个模型解释了动态类型：名字不携带类型信息，类型属于对象，`x = 1` 之后再 `x = "1"` 完全合法，只是把标签从整数挪到了字符串上。也解释了 `NameError`：名字从未被绑定过，查表找不到，与「值是空」是两回事。命名上用小写加下划线、让名字体现业务口径（如 `avg_speed_kmh`），比 `a`、`x1` 更能减少后续统计口径的笔误。

### 2.2.2 语法要点

| 写法 | 含义 | 说明 |
|---|---|---|
| `speed = 62` | 基本赋值 | `=` 是赋值，不是"等于"（等于用 `==`） |
| `a = b = 0` | 连续赋值 | 两个变量同时指向 0 |
| `a, b = 1, 2` | 多变量赋值 | 第 6 章讲函数多返回值时会大量使用 |
| `x += 1` | 复合赋值 | 等价 `x = x + 1`，另有 `-=` `*=` `/=` 等 |

**命名规范（教学统一要求）**：

| 规则 | 正确 | 错误 |
|---|---|---|
| 只用字母、数字、下划线 | `avg_gap`、`speed2` | `avg-gap`、`车速#1` |
| 不以数字开头 | `lane1` | `1lane` |
| 区分大小写 | `Speed` 与 `speed` 是两个变量 | —— |
| 见名知义，小写 + 下划线 | `avg_gap`、`flow_total` | `a`、`b`、`tmp1` |
| 不占用关键字 | `total`、`level` | `if`、`for`、`class`、`True` |

### 2.2.3 常见用法与易错点

- **`=` 与 `==` 混用**：`if speed = 60` 是语法错误，判断相等必须写 `==`（第 3 章）。
- **变量名用了中文或拼音全拼**：能运行，但不符合作业与工程规范；交通场景术语建议 `density`、`capacity`、`queue_len`。
- **忘记初始化累加变量**：第 3 章循环中 `total += flow` 之前必须有 `total = 0`。
- **误以为变量是"标签贴在值上"还是"盒子"**：本章按"盒子"理解即可；第 5 章处理列表时会看到"多个名字指向同一对象"的现象。

### 2.2.4 随堂示例：变量是贴了标签的盒子

**核心要点**：验证变量的三件事：用 `=` 赋值、用变量名取值、重新赋值会覆盖旧值；同时观察变量的命名要能看出含义。

````python
road = "京港澳 K12"      # 字符串变量：路段名
lanes = 3                # 整数变量：车道数
speed = 62.5             # 浮点变量：平均车速

print(road, "|", lanes, "车道 |", speed, "km/h")

lanes = 4                # 重新赋值：变量改指向新值 4
print("调整后车道数：", lanes)
print("通行能力：", lanes * 2200, "辆/h")
````

**预期输出**：

```text
京港澳 K12 | 3 车道 | 62.5 km/h
调整后车道数： 4
通行能力： 8800 辆/h
```

**变式代码（迁移练习）**：把随堂示例的「重新赋值」迁移为「读旧值 → 算新值 → 再绑定」的累加动作，并用带单位的命名降低歧义。

````python
# 赋值即创建：名字绑定到右边的对象，类型由对象决定
road = "机场北"
flow = 820
flow = flow + 180          # 先读旧值算出新对象，再把名字改绑过去
print("路段：", road, "| 更新后流量：", flow)
# 命名用小写加下划线，把单位写进名字，避免口径笔误
avg_speed_kmh = 62.5
print("平均车速：", avg_speed_kmh, "km/h")
````

**预期输出**：

```text
路段： 机场北 | 更新后流量： 1000
平均车速： 62.5 km/h
```

### 2.2.5 实操步骤（上机跟随）

1. 运行后确认输出 `1000`，说明 `flow = flow + 180` 先读到 820 再写回，而不是覆盖了等号左边。
2. 把 `+ 180` 改成 `+ 0`，观察输出变成 820，验证每次赋值都会重新绑定。
3. 在第一行后面加 `print(type(road))`，确认名字本身不声明类型，类型来自它绑定的对象。
4. 把 `road` 重新赋值为 `820` 再打印，观察类型随之变化，理解动态类型。
5. 把 `avg_speed_kmh` 改成 `s` 并读一遍代码，体会含单位命名对统计口径的帮助。

### 2.2.6 常见问题 FAQ

| 现象 | 根因 | 处理 |
|---|---|---|
| `NameError: name 'x' is not defined` | 名字从未被赋值绑定过 | 先赋值再使用；注意大小写拼写一致 |
| 重新赋值后别处的值也变了？ | 多个名字绑定了同一个**可变**对象（如列表） | 本章整数/字符串不可变；第 5 章会专门处理可变对象的引用问题 |
| `flow = flow + 1` 看不懂 | 把 `=` 当成了数学等号 | 它是赋值号：先算右边，再把结果绑定到左边 |
| 变量名用了中文或数字开头报语法错 | 标识符规则不允许 | 用字母或下划线开头，推荐小写加下划线 |

## 2.3 知识点 2.2　四种基本数据类型

### 2.3.1 核心概念

| 类型 | 关键字 | 字面量示例 | 交通场景 |
|---|---|---|---|
| 整数 | `int` | `62`、`0`、`-5` | 车道数、车辆数、相位数 |
| 浮点数 | `float` | `2.5`、`62.0`、`1.5e2` | 路段长度、密度、平均车速 |
| 字符串 | `str` | `"京A12345"`、`"2.5"` | 车牌、路段名、日志行 |
| 布尔值 | `bool` | `True`、`False` | 是否超速、是否饱和 |

`type(变量)` 返回该值的类型，是排查类型问题最直接的工具。

**原理补充（概念深化）**：类型决定的是**这个对象支持哪些运算、以及同一个运算符会做什么**，而不只是「长什么样」。`+` 作用在 `str` 上是拼接、作用在 `int` 上是加法、`int` 与 `float` 混合时结果自动升为 `float`（因为浮点能表示的集合更大，不会丢信息）。比较运算则永远返回 `bool`，`True` / `False` 在 Python 里是 `int` 的子类，所以能参与算术，但教学上不建议这么用。`type(x)` 返回的是**类型对象**而不是字符串，打印出来形如 `<class 'int'>`，这正是「类型属于对象」的直接证据。四类基本类型还对应四种业务语义：整数用于计数、浮点用于度量、字符串用于标识、布尔用于判定；把车牌写成整数、把车速写成字符串，是后续统计出错的高频源头。

### 2.3.2 语法要点

| 表达式 | 结果 | 说明 |
|---|---|---|
| `type(62)` | `<class 'int'>` | 整数字面量 |
| `type(2.5)` | `<class 'float'>` | 含小数点的字面量即 float |
| `type("62")` | `<class 'str'>` | 引号内一律是字符串，与内容无关 |
| `type(True)` | `<class 'bool'>` | 布尔值只有两个 |
| `type(2/1)` | `<class 'float'>` | `/` 的结果**永远是 float** |

**两条关键区分**：

- `62` 与 `"62"` 是**不同类型**：前者能参与数学运算，后者不能。
- `True` / `False` 首字母必须大写；`true` 会被当作未定义的变量名，报 `NameError`。

### 2.3.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| `"120" + 5` | `TypeError: can only concatenate str` | 先 `int("120")` 或改用 `eval(input())` |
| `"120" * 5` | 得到 `"120120120120120"`，**不报错** | 字符串 `*` 整数是重复，不是乘法 |
| `int("2.5")` | `ValueError: invalid literal for int()` | 先 `float("2.5")` 再取整 |
| `float("abc")` | `ValueError` | 输入校验（第 9 章 try-except） |
| 打印 `True` 期望"是" | 输出 `True` | 需求如此即正确；要中文须手写字符串 |

### 2.3.4 随堂示例：四种基本类型与 type()

**核心要点**：用 `type()` 逐个确认 `str` / `int` / `float` / `bool` 四种类型，并验证同一运算符在不同类型上的行为差异（数字相加是求和，字符串相加是拼接）。

````python
plate = "京A12345"       # str
lanes = 3                # int
speed = 62.5             # float
jam = True               # bool

print(type(plate), type(lanes), type(speed), type(jam))

# 同样的 + 号，类型不同结果完全不同
print("3 + 2 =", 3 + 2)
print("3.0 + 2 =", 3.0 + 2)
print("『京A』+『12345』=", "京A" + "12345")
````

**预期输出**：

```text
<class 'str'> <class 'int'> <class 'float'> <class 'bool'>
3 + 2 = 5
3.0 + 2 = 5.0
『京A』+『12345』= 京A12345
```

**变式代码（迁移练习）**：把随堂示例的「类型对照」迁移为「同一 `+` 号在四种类型上的不同行为」，用 `type()` 现场验证结果类型。

````python
# 同一个 + 号，作用于不同类型时含义完全不同
plate = "京A" + "12345"       # str + str 是拼接
lanes = 2 + 1                 # int + int 是加法
speed = 60 + 2.5              # int + float 结果升为 float
jam = 820 > 1500              # 比较运算结果是 bool
print(plate, type(plate))
print(lanes, type(lanes))
print(speed, type(speed))
print(jam, type(jam))
````

**预期输出**：

```text
京A12345 <class 'str'>
3 <class 'int'>
62.5 <class 'float'>
False <class 'bool'>
```

### 2.3.5 实操步骤（上机跟随）

1. 先预测四行各自的输出类型，再运行对照；重点看 `60 + 2.5` 为什么是 `float`。
2. 把 `lanes = 2 + 1` 改成 `2 + 1.0`，观察类型立刻变成 `float`，验证「向更宽的类型提升」。
3. 把 `jam` 的 `>` 改成 `<`，观察布尔值翻转但类型仍是 `bool`。
4. 试着写 `plate + lanes`，记录报错信息；思考为什么字符串与整数不能直接相加。
5. 用 `type(speed) == float` 打印一次，确认 `type()` 返回的是可比较的类型对象而非字符串。

### 2.3.6 常见问题 FAQ

| 现象 | 根因 | 处理 |
|---|---|---|
| `820` 加 `"60"` 报错 | str 与 int 之间没有定义 `+` | 先 `int("60")` 转换，或统一数据口径 |
| `type(x)` 打印出 `<class 'int'>` | 这是类型对象的正常表示 | 需要纯名字时可写 `type(x).__name__`，第 11 章会讲属性访问 |
| 小数计算结果有尾巴 | 二进制浮点无法精确表示多数十进制小数 | 第 2 章末用 `round()` 控制显示精度，不代表底层精度变了 |
| `True + 1` 居然等于 2 | `bool` 是 `int` 的子类 | 理解即可，业务代码里不要依赖这种隐式转换 |

## 2.4 知识点 2.3　输入与类型转换

### 2.4.1 核心概念

`input("提示语")` 在屏幕上显示提示并等待用户输入，**返回值永远是字符串**，即使用户输入的是 `2.5`。这是初学者第一个大坑：`lanes = input("车道数：")` 之后写 `2200 * lanes`，得到的是把字符串重复 2200 次。

解决办法有两条：

1. `eval(input("…"))`：把输入串**按 Python 表达式求值**，输入 `2.5` 得 float，输入 `3` 得 int，一次调用自动选对类型；
2. `int(input("…"))` / `float(input("…"))`：显式转换，意图更明确、更安全。

**原理补充（概念深化）**：`input()` 的返回值恒为 `str`，是因为终端只能交给程序**一串字符**，程序无法凭空知道用户想的是数字还是文本。因此类型转换不是可选项，而是所有输入处理的必经步骤。`int()` 只接受整数形态的串，`float()` 接受带小数点的串，`eval()` 则把字符串当 Python 表达式求值——它最省事也最危险：既能算出 `2.5`，也能执行任意表达式，真实项目里绝不应对外部输入使用。本章从工程角度更推荐 `int()` / `float()` 显式转换，因为意图清楚、失败时异常信息明确。还有一个易混点：转换函数**返回新对象**，不会原地修改原字符串；字符串本身不可变。把「读入 → 校验 → 转换 → 使用」固化为四步，后面第 9 章做录入容错、第 14 章解析 JSON 时都会复用。

### 2.4.2 语法要点

| 写法 | 输入 `3` 得到 | 输入 `2.5` 得到 | 适用 |
|---|---|---|---|
| `input()` | `"3"`（str） | `"2.5"`（str） | 车牌、路段名等文本 |
| `int(input())` | `3`（int） | 报 `ValueError` | 车道数、车辆数等整数 |
| `float(input())` | `3.0`（float） | `2.5`（float） | 长度、费率等可带小数的量 |
| `eval(input())` | `3`（int） | `2.5`（float） | 课堂快速演示，类型自动 |

**三个转换函数**：

| 函数 | 作用 | 示例 |
|---|---|---|
| `int(x)` | 转整数，**直接截断小数**（非四舍五入） | `int(2.9)` → `2` |
| `float(x)` | 转浮点数 | `float("2.5")` → `2.5` |
| `str(x)` | 转字符串，供拼接使用 | `str(130)` → `"130"` |

### 2.4.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 忘记转换 | 字符串重复（超长输出） | `eval(input())` 或 `int(input())` |
| 用 `int()` 读带小数的长度 | `ValueError` | 改用 `float()` 或 `eval()` |
| `print("周期：" + 130)` | `TypeError` | `str(130)`，或改用逗号：`print("周期：", 130)` |
| `eval()` 输入了非数字 | `NameError` / `SyntaxError` | 正式系统用 `float()` + 第 9 章异常处理 |
| 以为 `int()` 会四舍五入 | `int(2.9)` 得 `2` | 需要四舍五入用 `round(2.9)` |

> **安全提示**：`eval()` 会执行任意表达式，只适合教学与受控环境；真实系统接收外部输入时应用 `float()` 配合异常处理，这一点在第 9 章会再次强调。

### 2.4.4 随堂示例：input() 拿到的是字符串

**核心要点**：验证 `input()` 的返回值恒为 `str`，必须用 `int()` / `float()` / `eval()` 转换后才能参与算术；这里用字面量模拟用户输入，避免课堂上等待输入。

## 2.5 知识点 2.4　算术运算符

### 2.5.1 核心概念

| 运算符 | 名称 | 示例 | 结果 | 说明 |
|---|---|---|---|---|
| `+` | 加 | `820 + 760` | `1580` | 数字相加；字符串相加是拼接 |
| `-` | 减 | `60 - 15` | `45` | —— |
| `*` | 乘 | `2200 * 3` | `6600` | 字符串乘整数是重复 |
| `/` | 真除 | `7 / 2` | `3.5` | **结果永远是 float** |
| `//` | 整除（取商） | `7 // 2` | `3` | 向下取整，负数为 `-4` |
| `%` | 取余 | `7 % 2` | `1` | 求"不能整除的剩余" |
| `**` | 幂 | `2 ** 3` | `8` | 开方可写 `x ** 0.5` |

**优先级**：`**` > `*` `/` `//` `%` > `+` `-`，同级从左到右。不确定时加括号，可读性优先。

**原理补充（概念深化）**：`/`、`//`、`%` 三个运算符来自同一个整数除法恒等式：`a == (a // b) * b + (a % b)`。`/` 永远返回 `float`，哪怕能整除（`130 / 5` 得 `26.0`），这是 Python 3 相对 Python 2 的重要改变；`//` 是**向下取整**而非截断，所以负数 `-7 // 2` 得 `-4` 而不是 `-3`；`%` 的结果符号跟随除数，配合向下取整才能保证恒等式成立。信号配时里这三个运算各有位置：`/` 算平均、`//` 算整份基础时长、`%` 算余下的秒数或周期内的相位偏移。`**` 求幂、`+=` 这类增强赋值则是「先算再绑定」的简写，本质上与 `x = x + 1` 完全等价。理解恒等式后，很多「差一秒」的配时 bug 都能一眼定位。

### 2.5.2 语法要点：`//` 与 `%` 的组合用法

这是配时、均分、分组类问题的标准套路：

- `cycle // phases` 得到**每份的基础值**（商）；
- `cycle % phases` 得到**无法均分的余数**；
- 余数补给某一方，就得到不等分配。

典型案例：周期 130 秒分 4 个相位 → 每相位 32 秒，余 2 秒补主干道 → 主干道 34 秒、支路 32 秒。

### 2.5.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 该用 `/` 用了 `//` | 密度算成整数 12 而非 12.8 | 涉及"率、密度、均值"一律用 `/` |
| `10 / 0` | `ZeroDivisionError` | 计算前判断分母（第 3 章 if、第 9 章 try） |
| 优先级误判 | `2 + 3 * 4` 得 `14` 而非常 20 | 加括号：`(2 + 3) * 4` |
| `-7 // 2` | 得 `-4`（向下取整） | 负数场景改用 `int(-7 / 2)` 得 `-3` |
| 百分比少乘 100 | 0.78 显示成 0.78% | `soc / 100` 与 `soc * 100` 分清对象 |

### 2.5.4 随堂示例：// 与 % 拆分信号周期

**核心要点**：验证整除 `//`（取商的整数部分）与取余 `%`（取不够一份的余数）的配合用法，这是「平均分配 + 余数补给」这类业务的通用写法。

````python
cycle = 130          # 信号周期总时长（秒）
share = 4            # 主干道 : 支路 = 3 : 1，共 4 份

print("普通除法：", cycle / share)
print("整除：", cycle // share)
print("取余：", cycle % share)

base = cycle // share        # 每份基础时长：向下取整
rest = cycle % share         # 除不尽的余数：单独补给主干道
main_green = base * 3 + rest # 主干道拿 3 份，再吃下余数
side_green = base            # 支路拿 1 份
print("主干道绿灯：", main_green, "秒")
print("支路绿灯：", side_green, "秒")
print("合计：", main_green + side_green, "秒（正好等于周期）")
````

**预期输出**：

```text
普通除法： 32.5
整除： 32
取余： 2
主干道绿灯： 98 秒
支路绿灯： 32 秒
合计： 130 秒（正好等于周期）
```

**变式代码（迁移练习）**：把随堂示例的配时拆分迁移为「同一组数据跑四种除法类运算」，并用恒等式反算回原值做自检。

````python
# 同一组数据，四种除法类运算给出四个不同结果
cycle = 130
phases = 4
print("真除 /：", cycle / phases, type(cycle / phases))
print("整除 //：", cycle // phases)
print("取余 %：", cycle % phases)
# 整除取商 + 取余，可以还原被除数
print("还原：", cycle // phases * phases + cycle % phases)
print("幂 **：", 2 ** 5)
````

**预期输出**：

```text
真除 /： 32.5 <class 'float'>
整除 //： 32
取余 %： 2
还原： 130
幂 **： 32
```

### 2.5.5 实操步骤（上机跟随）

1. 运行后确认 `/` 的结果是 `32.5` 且类型为 `float`，而 `//` 是 `32`。
2. 核对最后第二行是否恰好等于 130，验证恒等式 `a == (a // b) * b + a % b`。
3. 把 `cycle` 改成 `-7`、`phases` 改成 `2`，观察 `//` 得 `-4`，理解向下取整不同于截断。
4. 把 `2 ** 5` 改成 `2 ** 0.5`，观察幂运算结果变成浮点。
5. 把 `cycle // phases * phases` 的乘号误写成加号，看还原值如何偏离，体会运算优先级。

### 2.5.6 常见问题 FAQ

| 现象 | 根因 | 处理 |
|---|---|---|
| `130 / 5` 得到 `26.0` 而不是 `26` | Python 3 的 `/` 恒返回 float | 需要整数商就用 `//` |
| 负数整除结果少 1 | `//` 是向下取整，不是向零截断 | 按数轴左侧方向理解，或先取绝对值再处理 |
| 配时总秒数差 1—2 秒 | 各相位都用 `//` 取整，余数被丢掉 | 把余数补给主干道相位，或统一用 `/` 后再 `round` |
| `+=` 与 `= +` 混淆 | 后者是赋值加一元正号 | 增强赋值必须连写 `+=`，中间不能有空格拆开 |

## 2.6 知识点 2.5　比较运算符与逻辑运算符

### 2.6.1 核心概念

比较运算的结果是布尔值（`True` / `False`），逻辑运算把多个布尔值组合成一个结论。本章虽不写 `if`，但先把"结论算出来并打印"是第 3 章判断的预备动作。

| 比较运算符 | 含义 | 示例 | 结果 |
|---|---|---|---|
| `>` `<` | 大于 / 小于 | `62 > 60` | `True` |
| `>=` `<=` | 大于等于 / 小于等于 | `1580 <= 1500` | `False` |
| `==` | 等于 | `62 == 62.0` | `True`（数值比较不看类型） |
| `!=` | 不等于 | `"畅通" != "拥堵"` | `True` |

| 逻辑运算符 | 含义 | 说明 |
|---|---|---|
| `and` | 与 | 两侧都为真才为真 |
| `or` | 或 | 任一侧为真即为真 |
| `not` | 非 | 取反 |

**链式比较**：`60 <= cycle <= 180` 与 `cycle >= 60 and cycle <= 180` 完全等价，前者更接近数学写法。

**原理补充（概念深化）**：比较运算先各自算出一个 `bool`，`and` / `or` / `not` 再把布尔值组合起来。这里有两个容易被忽略的机制：一是**短路求值**，`a and b` 在 `a` 为假时根本不会计算 `b`，`a or b` 在 `a` 为真时也不会算 `b`，第 9 章做容错时会利用这一点避免对坏数据取值；二是 `==` 比较的是**值**，`62 == 62.0` 为真，而 `is` 比较的是**是否为同一对象**，两者不能混用。多个比较还能连写成 `20 <= speed < 60`，等价于 `20 <= speed and speed < 60`，但只求值一次。本章虽然还没写 `if`，但把结论先算出来打印，正是第 3 章分支判断的预备动作：分支只是拿这个 `bool` 决定走哪条路，不再做第二次计算。

### 2.6.2 语法要点

- 比较运算的优先级**高于** `and` / `or`，因此 `speed >= 60 and density < 10` 无需额外括号。
- `not` 的优先级高于 `and` / `or`，`not a and b` 意为 `(not a) and b`。
- `and` / `or` 有"短路"特性：`False and X` 不再计算 X；`True or X` 同理。第 9 章可利用它避免除零。

### 2.6.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| `=` 当 `==` 用 | `SyntaxError` | 判断相等写 `==` |
| `"60" > 50` | `TypeError` | 先转换类型 |
| 用 `and` 连接两个区间 | `speed >= 60 and <= 40` 语法错误 | 右侧写完整：`speed >= 60 and speed <= 40` |
| 混淆 `and` 与"并且包含" | 逻辑理解错 | `and` 是"两个条件同时成立" |

### 2.6.4 随堂示例：比较与逻辑运算产出布尔值

**核心要点**：验证比较运算返回 `bool`，再用 `and` / `or` / `not` 组合成复合条件——这正是第 3 章 `if` 语句的判断依据。

````python
speed = 45           # 车速（km/h）
density = 8          # 车流密度（辆/(km·车道)）

print("车速 ≥ 60：", speed >= 60)              # False
print("密度 < 10：", density < 10)              # True
print("两个条件同时满足：", speed >= 60 and density < 10)   # False
print("任一条件满足：", speed >= 60 or density < 10)        # True
print("对结果取反：", not (speed >= 60))        # True
print("类型是：", type(speed >= 60))
````

**预期输出**：

```text
车速 ≥ 60： False
密度 < 10： True
两个条件同时满足： False
任一条件满足： True
对结果取反： True
类型是： <class 'bool'>
```

**变式代码（迁移练习）**：把随堂示例的「直接打印组合结论」迁移为「先算两个独立布尔量，再用三种逻辑运算组合」，顺带验证数值比较不看类型。

````python
# 先算两个独立的比较结论，再用 and / or / not 组合
speed = 52
flow = 1680
a = speed >= 60
b = flow > 1500
print("车速达标：", a, "| 流量超限：", b)
print("且 and：", a and b, "| 或 or：", a or b)
print("非 not：", not a)
# == 比较数值不看类型；连写比较等价于两个条件相与
print("62 == 62.0：", 62 == 62.0)
print("区间判断：", 40 <= speed < 60)
````

**预期输出**：

```text
车速达标： False | 流量超限： True
且 and： False | 或 or： True
非 not： True
62 == 62.0： True
区间判断： True
```

### 2.6.5 实操步骤（上机跟随）

1. 先算出 `a`、`b` 的预期值再运行，确认分别是 `False` 与 `True`。
2. 核对 `a and b` 为假、`a or b` 为真，说出短路求值在这里跳过了哪一步。
3. 把 `speed` 改成 `72`，预测五个输出各自的变化，再运行验证。
4. 把 `40 <= speed < 60` 拆成 `40 <= speed and speed < 60`，确认结果一致。
5. 思考 `==` 与 `is` 的差别：本章只用 `==` 比较数值，不要提前引入身份判断。

### 2.6.6 常见问题 FAQ

| 现象 | 根因 | 处理 |
|---|---|---|
| `and` / `or` 与普通加减混写结果意外 | 逻辑运算优先级低于比较、高于赋值 | 拿不准就加括号，或先把结论存进变量 |
| `62 == 62.0` 居然为真 | `==` 比较数值，跨 int/float 也成立 | 需要连类型一起判断时才另做 `type()` 比较 |
| 用 `=` 写条件报语法错 | 单个等号是赋值，比较要用 `==` | 检查所有条件里的等号是否成对 |
| 短路导致某个函数没被调用 | `and` / `or` 会跳过不必要的右侧求值 | 把必须执行的动作放在逻辑表达式之外 |

## 2.7 知识点 2.6　`round()` 与输出精度

### 2.7.1 核心概念

`round(x, n)` 返回**保留 n 位小数的新值**。关键认识：它**不改变原变量**，只是把"四舍五入后的结果"交给你使用；原变量的精度不变。

**原理补充（概念深化）**：`round(x, n)` 返回的是一个**新数值**，原变量一丝未动；它解决的是「显示成几位」，不是「内存里存几位」。因此连续两次 `round` 同一变量不会累积误差，但若把 `round` 的结果再赋回原变量，就会真的丢精度。第二个要点是 Python 采用**银行家舍入**（ties to even）：恰好落在中点时取最接近的偶数，`round(2.5)` 得 2、`round(3.5)` 得 4，这与很多教材说的「四舍五入」不同。第三，浮点数在二进制里无法精确表示多数十进制小数，`2.675` 实际略小于 2.675，所以 `round(2.675, 2)` 得 2.67 而非 2.68。`n` 可以是负数，表示对十位、百位取整，且结果会带 `.0` 变成浮点。财务、计费类计算不应直接依赖浮点 `round`，教学报表里用它控制显示即可。

### 2.7.2 语法要点

| 写法 | 结果 | 说明 |
|---|---|---|
| `round(12.778, 2)` | `12.78` | 保留 2 位 |
| `round(26.0416, 1)` | `26.0` | 保留 1 位，末尾补 0 |
| `round(26.6)` | `27` | 省略第二参数 → 取整为 **int** |
| `round(2.675, 2)` | `2.67` | 二进制浮点误差导致，非四舍五入失效 |

### 2.7.3 常见用法与易错点

| 易错点 | 说明 | 建议 |
|---|---|---|
| 以为 `round` 改了原变量 | `round(x, 2)` 后 `x` 不变 | 需要保留结果就 `x = round(x, 2)` |
| 用 `round` 做金额累加 | 多次舍入误差累积 | 金额按"分"存整数，最后再除 100 |
| 期望 `round(2.675, 2)` 得 2.68 | 浮点数二进制表示所致 | 需要精确十进制用 `decimal` 模块 |
| 想让输出"总是两位小数" | `round(12.8, 2)` 显示 `12.8` 而非 `12.80` | 用第 4 章 `f"{x:.2f}"` 控制显示 |

---

### 2.7.4 随堂示例：round() 控制有效位数

**核心要点**：验证 `round(x, n)` 的第二参数可以是正数（保留小数位）、0（取整）、负数（对整数位取整），并注意浮点误差导致个别结果与直觉不同。

````python
density = 96 / (2.5 * 3)
print("原始值：", density)
print("保留 2 位：", round(density, 2))
print("保留 0 位：", round(density))
print("对十位取整：", round(density, -1))

# 浮点误差：2.675 在内存中略小于 2.675
print("round(2.675, 2) =", round(2.675, 2))
````

**预期输出**：

````python
# round(x, n) 返回新值，原变量精度不变
density = 96 / 7
print("原始：", density)
print("保留 2 位：", round(density, 2))
print("原变量仍是：", density)
# 负数 n 表示对十位、百位取整，结果带 .0
print("对十位取整：", round(density, -1))
# 恰好在中点时取偶数（银行家舍入）
print("round(2.5)：", round(2.5), "| round(3.5)：", round(3.5))
````

### 2.7.6 常见问题 FAQ

### 任务情境

某十字路口信号周期总时长按相位数平均分配，不能整除的余数补给主干道方向。需要算出各相位绿灯时长，并校验周期是否落在合理区间（60~180 秒）。

### 任务要求

1. 用 `//` 求每相位基础时长，用 `%` 求余数；
2. 用复合赋值 `+=` 把余数补给主干道；
3. 用比较运算符与 `and` 输出周期是否合理的布尔值；
4. 用 `str()` 把数字转成字符串参与拼接输出。

### 参考代码

````python
# ---------- 第 02 章 变量与数据类型 · 案例 2-2：信号配时周期拆分 ----------
# 对应知识点：2.3 str() 类型转换、2.4 // 与 % 及复合赋值、2.5 比较与逻辑运算、2.6 输出拼接
# 功能说明：把信号周期按相位平均分配，余数补给主干道，并判断周期是否落在 60~180 秒合理区间

# 信号周期按相位平均分配，余数补给主干道
cycle = eval(input("信号周期总时长(s)："))
phases = eval(input("相位数："))

base = cycle // phases        # 每相位基础绿灯
rest = cycle % phases         # 不能整除的余数

print("每相位基础绿灯：", base, "s")
print("余数：", rest, "s")

base += rest                  # 复合赋值：主干道多拿余数
print("主干道绿灯：", base, "s")
print("支路绿灯：", cycle // phases, "s")

# 比较运算 + 逻辑运算，结果为 bool
print("周期是否合理(60~180s)：", cycle >= 60 and cycle <= 180)
print("是否严重超长(>180s)：", cycle > 180)

# str() 把数字转换成字符串才能与文本拼接
print("本周期设定为 " + str(cycle) + " 秒，共 " + str(phases) + " 个相位")
````
