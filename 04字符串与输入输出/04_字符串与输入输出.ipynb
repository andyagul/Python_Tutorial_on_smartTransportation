## 4.2 知识点 4.1　索引、切片与 `len()`

### 4.2.1 核心概念

字符串中每个字符都有一个编号，称为**下标（索引）**，**从 0 开始**。`plate[0]` 是第 1 个字符，`plate[1]` 是第 2 个。

**负索引**从末尾数：`plate[-1]` 是最后一个字符，`plate[-2]` 是倒数第二个。

**切片** `s[start:end:step]` 取一段子串，规则是**左闭右开**：包含 `start`，不包含 `end`。

**原理补充（概念深化）**：Python 3 的字符串是 Unicode 码点序列，每个字符占一个下标，所以 `len()` 按字符算，一个汉字是一而不是二。这与按字节计数、显示列宽是两回事，初学者最易混淆三者。索引会做边界检查，取越界位置立即抛异常；切片则先把越界端点夹逼回合法范围再取子串，所以 `s[0:100]` 也不会报错，两者底层并非同一套逻辑。切片新建字符串对象，原串因不可变不受影响；负下标在内部被换算成长度减该数。左闭右开让 `[i:j]` 长度恰为终点减起点，与区间函数一致，是序列协议统一的约定。

### 4.2.2 语法要点

| 写法 | 含义 | `s = "京A12345"` 的结果 |
|---|---|---|
| `s[0]` | 第 1 个字符 | `京` |
| `s[1]` | 第 2 个字符 | `A` |
| `s[-1]` | 最后一个字符 | `5` |
| `s[2:]` | 从第 3 个到末尾 | `12345` |
| `s[:2]` | 前 2 个字符 | `京A` |
| `s[2:5]` | 第 3—5 个 | `123` |
| `s[::2]` | 每隔一个取一个 | `京130` |
| `s[::-1]` | 倒序 | `54321A京` |
| `len(s)` | 字符个数 | `7` |

### 4.2.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 认为下标从 1 开始 | `s[7]` 越界 | 下标 0 起，最大为 `len(s)-1` |
| 切片右端写错 | `s[0:2]` 只得 2 个字符 | 要含第 2 个字符须写 `s[0:3]` 或 `s[:3]` |
| 越界访问 | `IndexError: string index out of range` | 先判 `len()`，或用 `s[2:]` 这类安全切片 |
| 以为切片越界会报错 | `s[0:100]` 正常返回全串 | 切片不报错，索引才报错 |
| 中文字符按字节数 | `len("京")` 得 1（不是 2） | Python 3 按**字符**计数 |

### 4.2.4 随堂示例：用索引与切片拆车牌

**核心要点**：验证字符串下标从 0 开始、负数下标从末尾数、切片 `[start:stop]` 不含 `stop`，并用 `len()` 判断长度。

````python
plate = "京A12345"

print("长度：", len(plate))
print("第 1 位（下标 0）：", plate[0])
print("最后 1 位（下标 -1）：", plate[-1])
print("发牌机关（下标 1）：", plate[1])
print("号牌序号（2 到末尾）：", plate[2:])
print("前 3 位：", plate[:3])
print("后 3 位：", plate[-3:])
````

```text

第 1 位（下标 0）： 京

```

### 4.2.6 常见问题 FAQ

| s[::-1] 得到反串但原串没变 | 切片返回新字符串对象，字符串不可变 | 倒序结果赋给新变量，如 rev = s[::-1] |

## 4.3 知识点 4.2　数据清洗方法

### 4.3.1 核心概念

现实数据常带空格、大小写不一、含多余字符。**清洗**就是把数据变成统一格式。常用四个方法：

| 方法 | 作用 | 示例 | 结果 |
|---|---|---|---|
| `strip()` | 去掉首尾空白（空格、`\t`、`\n`） | `"  京A12345  ".strip()` | `"京A12345"` |
| `upper()` | 全部转大写 | `"京a12345".upper()` | `"京A12345"` |
| `lower()` | 全部转小写 | `"ABC".lower()` | `"abc"` |
| `replace(旧, 新)` | 替换全部匹配子串 | `"车速 62".replace("车速", "speed")` | `"speed 62"` |

**关键理解**：字符串**不可变**，这些方法都返回**新字符串**，原变量不变。因此必须写 `plate = plate.strip()` 或串接 `raw.strip().upper()`。

**原理补充（概念深化）**：这几个清洗方法看似在改字符串，实则一个都不改：字符串创建后内容不可变，每次清洗都造一个新字符串对象再返回。所以不接收返回值时，新对象立刻被丢弃，原变量仍是带空格的旧值；这也是链式调用能成立的前提，点号左侧每次拿到的都是上一方法新建的对象。`strip()` 只扫描首尾、遇第一个非空白字符即停，中间空格一律不动，`replace()` 才做全局匹配并可用第三参数限定次数。大小写转换按统一映射进行，对汉字无效，先统一再比较可避免小写与大写被判为不等。理解不可变加返回新值，就懂为何必须重新赋值。

### 4.3.2 语法要点

- 方法可以**链式调用**：`raw.strip().upper()` 先清洗再去空白，从左到右依次执行。
- `strip()` 只去**两端**，中间空格不动；要去掉所有空格用 `replace(" ", "")`。
- `replace()` 默认替换全部；可传第三参数限制次数 `replace("a", "b", 1)`。
- `upper()` / `lower()` 对中文无影响，只作用于英文字母。

### 4.3.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 忘记接收返回值 | `raw.strip()` 后 `raw` 仍有空格 | `raw = raw.strip()` |
| 用 `strip("京")` 想删中间字符 | 只删两端 | 用 `replace()` |
| 大小写混判 | `"a" == "A"` 为 `False` | 比较前统一 `upper()` |
| 清洗顺序混乱 | 先大写再去空格也可，但要有固定顺序 | 统一"去空白 → 统一大小写" |

### 4.3.4 随堂示例：清洗一条带空白的车牌

**核心要点**：验证 `strip()` 只去首尾空白、`upper()` 统一大小写、`replace()` 替换内容；三者可以链式调用，且都不改变原字符串。

````python
raw = "  京a12345  "          # 设备上报时带空格、字母小写
clean = raw.strip().upper()   # 先清洗再去大小写差异

print("原始：[" + raw + "]")
print("清洗：[" + clean + "]")
print("长度：", len(raw), "→", len(clean))
print("替换后：", clean.replace("京A", "京B"))
print("原字符串是否被改动：[" + raw + "]")     # 仍是原样，字符串不可变
````

**预期输出**：

```text
原始：[  京a12345  ]
清洗：[京A12345]
长度： 11 → 7
替换后： 京B12345
原字符串是否被改动：[  京a12345  ]
```

**变式代码（迁移练习）**：把示例的单个车牌换成带首尾空格、小写字母与过时编号的路段名，验证链式清洗顺序、`replace` 限定次数与中间空格不被 strip 去掉，且原串始终未变。

````python
# 门架上报路段名：首尾多空格、字母小写、编号需更新，演示链式清洗
raw = "  京港澳-k12  "
clean = raw.strip().upper().replace("K12", "K18")
print("原始：[" + raw + "]")
print("清洗后：[" + clean + "]")
print("原始长度：", len(raw), "→ 清洗后：", len(clean))
print("中间空格保留：[" + "京 港 澳".strip() + "]")
print("限定替换一次：", "K12-K12".replace("K12", "K18", 1))
print("原串不变：[" + raw.strip().upper() + "]")
````

**预期输出**：

```text

清洗后：[京港澳-K18]
原始长度： 11 → 清洗后： 7
中间空格保留：[京 港 澳]
限定替换一次： K18-K12
原串不变：[京港澳-K12]
```

### 4.3.5 实操步骤（上机跟随）

1. 先读『原始：[  京港澳-k12  ]』，确认首尾各两个空格、字母 k 小写、编号为 k12。
2. 看『清洗后：[京港澳-K18]』，验证 strip、upper、replace 三者从左到右依次执行。
3. 核对『原始长度：11 → 清洗后：7』，确认 strip 只删首尾共 4 个空格、中间内容未动。

5. 看末行『原串不变：[  京港澳-k12  ]』，确认所有方法都返回新串、必须重新赋值才生效。

### 4.3.6 常见问题 FAQ

| 现象 | 根因 | 处理 |
|---|---|---|
| raw.strip() 之后 raw 仍带空格 | 字符串不可变，strip 返回新串、原变量未变 | 写 raw = raw.strip() 或串接 raw.strip().upper() |
| strip('京') 想删中间汉字却没删掉 | strip 只扫描首尾，遇第一个非空白字符即停 | 删中间字符用 replace('京', '') |

| replace 把两处 K12 都换了 | 未传第三参数时默认替换全部匹配 | 只换一处写 replace('K12', 'K18', 1) |

## 4.4 知识点 4.3　查找与判断

### 4.4.1 核心概念

判断"某段文字是否在字符串里、在哪、出现几次"，用下面五个方法：

| 方法/运算符 | 返回 | 未找到时 | 交通场景 |
|---|---|---|---|
| `"京" in name` | `True` / `False` | `False` | 是否属于"京"字头高速 |
| `name.find("京")` | 首次出现的下标 | `-1` | 需要知道位置 |
| `name.startswith("京")` | `True` / `False` | —— | 前缀判断（省份简称） |
| `name.endswith("高速")` | `True` / `False` | —— | 后缀判断（道路类型） |
| `name.count("京")` | 出现次数（整数） | `0` | 统计关键词频次 |

**原理补充（概念深化）**：`in` 与 `find()` 的分工源于返回类型：前者只回答有没有、命中即短路返回布尔值；后者要给出首次下标，找不到时按约定返回 `-1` 这个哨兵值而非抛异常，好处是免于异常捕获，坏处是 `-1` 在布尔上下文为真、而位置 `0` 为假，于是把定位结果直接当条件用，会把恰好出现在开头误判成没找到。前后缀判断只比对首尾，计数按不重叠子串进行、传空串会得到长度加一。字符属性判断里汉字不属于单字节字符集，案例借这一结果反推首字符是汉字。

### 4.4.2 字符判断方法

| 方法 | 判断内容 | `"123".isalnum()` | `"京".isascii()` |
|---|---|---|---|
| `isalnum()` | 是否全为字母或数字 | `True` | —— |
| `isdigit()` | 是否全为数字 | `True` | —— |
| `isalpha()` | 是否全为字母 | `False`（含数字） | —— |
| `isascii()` | 是否全为 ASCII 字符 | —— | `False`（汉字非 ASCII） |

> `not prov.isascii()` 是判断"首字符是否为汉字"的简便写法，案例 4-1 用到。

### 4.4.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 用 `find()` 做布尔判断 | `if name.find(k)` 在位置 0 时判为假 | 用 `in`，或写 `!= -1` |
| 认为 `find` 找不到抛异常 | 实际返回 `-1` | 用 `-1` 判断 |
| `count("")` 结果异常 | 返回 `len(s)+1` | 空串不要用于 `count` |

## 4.5 知识点 4.4　`split()` 与 `join()`

### 4.4.1 核心概念

`split(分隔符)` 把字符串按分隔符**切成多段**；`join()` 把多段**拼回**一个字符串。二者是相反操作，也是日志处理与 CSV 处理的基础。

| 操作 | 写法 | 结果 |
|---|---|---|
| 切分 | `"a\|b\|c".split("\|")` | `["a", "b", "c"]` |
| 按空白切分 | `"a  b   c".split()` | `["a", "b", "c"]` |
| 拼接 | `"-".join(["a","b"])` | `"a-b"` |
| 重复 | `"=" * 40` | 40 个等号 |

**原理补充（概念深化）**：切分方法 `split()` 从左到右扫描、丢弃分隔符，把之间的子串依次收进一个新列表，因此段数等于分隔符出现次数加一，这条恒等式可反过来校验日志字段是否完整；不传参数时按任意连续空白切分并自动丢掉空段。拼接方法 `join()` 恰是逆操作：它由分隔符字符串调用、参数是字符串序列，内部先算总长一次性分配内存再写回，比循环里反复相加高效得多，这也是分隔符在点号左边的原因，因为它是字符串的方法而非列表的方法。两者都不改动原串，切分产新列表、拼接产新字符串。段内残留空格来自分隔符两侧，须逐段去除。

### 4.5.2 语法要点

- `split()` 不传参数时按**任意空白**切分，并自动忽略连续空格，是清洗文本最省事的写法。
- `split("|")` 切出的段数 = 分隔符数 + 1，可用它做校验（案例 4-2、练习 4-1 都用到）。
- `join()` 作用于**字符串**（分隔符），参数是"字符串的可迭代对象"；列表中若含数字须先 `map(str, ...)`（第 8 章）。
- `"=" * 40` 是字符串重复，第 2 章提过；比手写 40 个等号更不易出错。

### 4.5.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 分不清谁调 `join` | `["a","b"].join("-")` 报错 | 分隔符调 `join`：`"-".join([...])` |
| 分隔符含空格 | `split(" \| ")` 与 `split("\|")` 结果不同 | 先 `strip()` 再用 `split("\|")`，段内再 `strip()` |
| 段数校验漏掉 | 少一个字段就 `IndexError` | `len(parts)` 与预期比较（第 9 章还会用 try） |

### 4.5.4 随堂示例：split() 拆字段、join() 拼回去

**核心要点**：验证 `split(分隔符)` 得到列表（分隔符本身不保留）、`join()` 用指定字符把列表拼成字符串，二者互为逆操作。

````python
line = "08:15:32|京A12345|62|2"

parts = line.split("|")
print("切分结果：", parts)
print("字段数：", len(parts))
print("第 2 个字段（下标 1）：", parts[1])

print("前两段用 - 连接：", "-".join(parts[0:2]))
print("全部用逗号连接：", ",".join(parts))
````

**预期输出**：

## 4.6 知识点 4.5　三种格式化与对齐精度

### 4.6.1 核心概念

同一份数据有三种写法，f-string 是当前推荐：

| 写法 | 示例 | 说明 |
|---|---|---|
| `%` 格式化 | `"%s / %.1f km/h" % (road, speed)` | 老式，C 语言风格，读旧代码会遇到 |
| `format()` | `"{0} / {1:.1f}".format(road, speed)` | Python 3.0 引入，比 `%` 灵活 |
| **f-string** | `f"{road} / {speed:.1f}"` | Python 3.6+，变量直接嵌入，**本章推荐** |

**原理补充（概念深化）**：百分号、`format()` 与 f 字符串是同一套格式说明符的三代外壳：老式写法源自 C 语言，中代用花括号占位，新式在编译期就把花括号里的表达式编译成求值字节码、运行期不再解析模板，因而最快也最易读，这是本章推荐它的底层原因。冒号后的说明符交给类型自身的格式化协议，如 `:.1f` 控制小数位：对齐按字符数补空格而非显示列宽，逗号插入千分位。汉字按两个字符计宽却在等宽字体里占四列，中英混排必然参差，这是字符宽与终端列宽的错位，不是对齐写错。漏写 `f` 前缀时花括号只是普通字面量。

### 4.6.2 格式说明符（重点）

`f"{值:说明符}"` 中冒号后的部分控制显示格式：

| 说明符 | 含义 | 示例 | 结果（值 28.0） |
|---|---|---|---|
| `:<10` | 左对齐，占 10 字符 | `f"{'车速':<10}"` | `车速        ` |
| `:>10` | 右对齐，占 10 字符 | `f"{28:>10}"` | `        28` |
| `:^10` | 居中，占 10 字符 | `f"{'合计':^10}"` | `   合计   ` |
| `:.1f` | 保留 1 位小数 | `f"{28:.1f}"` | `28.0` |
| `:>10.2f` | 宽 10、右对齐、2 位小数 | `f"{28:>10.2f}"` | `     28.00` |
| `:,` | 千分位分隔 | `f"{10110:,}"` | `10,110` |

### 4.6.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 忘记 `f` 前缀 | 输出原样 `{road}` | 字符串前加 `f` |
| 宽度写反 | 表头与数据错位 | 同一列**表头与数据用相同宽度** |
| 中英文混排错位 | 汉字占两列导致参差 | 见下方说明 |
| `%` 与 `%` 冲突 | 文本含百分号需写 `%%` | 或改用 f-string |

> **中英文混排的现实限制**：`f-string` 的宽度按**字符数**计算，而等宽字体下一个汉字占**两列**。因此宽度设为 10 时，"京港澳高速 K12"（9 个字符，显示宽度 13 列）会比其他行宽。实际项目的解决办法是：表头统一用中文、数据列用固定宽度、或改用表格控件（第 8 章 CSV、第 13 章图形界面）。

### 4.6.4 随堂示例：三种格式化写法的对齐效果

**核心要点**：验证同一张两列表格用 `f-string`、`%`、`format()` 三种写法都能实现；其中 `:<n` 左对齐补空格、`:>n` 右对齐、`:.1f` 控制小数位。

````python
road = "京开"
speed = 43.5

# 1) f-string：宽度与精度都写在花括号里
print(f"{'路段':<8}{'车速':>8}")
print(f"{road:<8}{speed:>8.1f}")

# 2) 百分号写法：类型符 %s / %.1f
print("百分号：%s / %.1f km/h" % (road, speed))

# 3) format() 写法：用 {} 占位
print("format：{} / {:.1f} km/h".format(road, speed))
````

**预期输出**：

## 案例 4-1　车牌号解析与合法性初筛

- **难度**：基础　**建议用时**：25 分钟

### 任务情境

卡口相机识别到车牌后，需要拆出省份简称、发牌机关代号与号牌序号，并按长度规则做一次合法性初筛（普通号牌 7 位、新能源号牌 8 位）。

### 任务要求

1. 输入车牌号，用 `strip()` 去掉首尾空格，用 `upper()` 统一大写；
2. 用索引取出第 1 位与第 2 位，用切片取出第 3 位之后的序号；
3. 用 `len()` 判断长度是 7 位还是 8 位，输出车牌类型；
4. 用 `isalnum()` 判断序号部分是否只含字母与数字。

### 输入规格

| 项 | 要求 |
|---|---|
| 数据来源 | 1 次 `input()` 读入原始车牌，示例 `京a12345 `（首尾有空格） |
| 类型 | `input()` 恒返回 `str`，不转数字、不参与算术 |
| 取值范围 | 规范化后 7 或 8 位；第 1 位汉字简称，第 2 位字母，其余 5—6 位字母或数字 |
| 非法值示例 | 只输空格触发 `IndexError`；`京A-2345` 使 `isalnum()` 为 `False`；6 位车牌走长度异常分支 |

### 输出规格

| 项 | 要求 |
|---|---|
| 行数 | 恰好 9 行：1 行输入回显、8 行解析结果 |
| 逐行格式 | 依次为规范化结果、字符总数、省份简称、发牌机关、号牌序号、车牌类型、序号是否只含字母数字、首字符是否为汉字 8 行 |

````python
# ---------- 第 04 章 字符串与输入输出 · 案例 4-1：车牌号解析与合法性初筛 ----------
# 对应知识点：4.1 索引 / 切片 / len()、4.2 strip() 与 upper()、4.3 isalnum() / isascii()、3.1 分支判定
# 功能说明：解析车牌字符串，取出省份简称、发牌机关与号牌序号，并按长度初筛车牌类型

# 车牌号解析与合法性初筛
raw = input("请输入车牌号：")
plate = raw.strip().upper()          # 去空白 + 统一大写

print("规范化结果：", plate)
print("字符总数：", len(plate))

prov = plate[0]         # 索引 0：省份简称
letter = plate[1]       # 索引 1：发牌机关代号
number = plate[2:]      # 切片：从第 3 位到末尾的序号

print("省份简称：", prov)
print("发牌机关：", letter)
print("号牌序号：", number)

if len(plate) == 7:
    print("车牌类型：普通民用号牌")
elif len(plate) == 8:
    print("车牌类型：新能源号牌")
else:
    print("车牌类型：长度异常，需人工复核")

print("序号是否只含字母数字：", number.isalnum())
print("首字符是否为汉字：", not prov.isascii())
````

````python
# ---------- 第 04 章 字符串与输入输出 · 案例 4-2：路况播报文本排版 ----------
# 对应知识点：4.5 f-string / % / format() 三种格式化与对齐精度、4.4 字符串重复拼接、3.1 状态判定
# 功能说明：把路段名、平均车速与排队长度排版为三列对齐报表，并按状态给出通行建议

road = input("路段名称：")
speed = eval(input("平均车速(km/h)："))
queue = eval(input("排队长度(m)："))

# 1) 表头三列对齐（左对齐 10、右对齐 10、右对齐 10）
print(f"{'路段':<10}{'车速(km/h)':>14}{'排队(m)':>12}")

# 2) 数据行：同一套宽度，保证纵向对齐
print(f"{road:<10}{speed:>14.1f}{queue:>12}")

# 3) 老式 % 格式化写法（对照用）
print("百分号写法：%s / %.1f km/h / %d m" % (road, speed, queue))

# 4) format() 写法（对照用）
print("format 写法：{0} / {1:.1f} km/h / {2} m".format(road, speed, queue))

if queue >= 300:
    status = "拥堵"
    tip = "建议绕行辅道"
elif queue >= 100:
    status = "缓行"
    tip = "保持车距，谨慎变道"
else:
    status = "畅通"
    tip = "正常通行"

print("=" * 40)
print(f"【路况播报】{road}　当前{status}")
print(f"平均车速 {speed:.1f} km/h，排队长度 {queue} m。{tip}。")
````

### 参考代码

````python
# ---------- 第 04 章 字符串与输入输出 · 练习 4-1 参考答案：车载日志行清洗与字段提取 ----------
# 对应知识点：4.2 strip() 清洗、4.4 count() 与 split()、4.3 replace()
# 功能说明：清洗一条车载日志，按竖线切分字段，校验分隔符数量并输出英文化文本

line = "  2026-10-29 08:15:32 | 京A12345 | 车速 62 | 车道 2  "

clean = line.strip()                 # 去首尾空白
print("清洗后：", clean)
print("分隔符个数：", clean.count("|"))

parts = clean.split("|")             # 按竖线切分
print("字段个数：", len(parts))
print("分隔符校验通过：", clean.count("|") == len(parts) - 1)

for i in range(len(parts)):
    print("字段" + str(i) + "：", parts[i].strip())

print("英文化输出：", clean.replace("车速", "speed").replace("车道", "lane"))
````

### 运行结果

```text
字段0： 2026-10-29 08:15:32

```

### 解析

1. **两级清洗**：整行 `strip()` 只去掉首尾空格，字段之间的空格仍在，因此逐字段还要 `strip()`。这是日志处理的通用套路，第 8 章读文件时会反复使用。
2. **字段数为 4 的由来**：3 个分隔符切出 4 段，公式是 `len(parts) == count("|") + 1`，题目要求校验的正是这个关系。
3. **`range(len(parts))` 的必要性**：本章尚未学 `for p in parts` 这种直接遍历容器的写法（第 5 章），只能用下标访问，因此要靠 `len()` 控制循环次数。
4. **链式 `replace`**：两次替换串接，得到英文字段，便于下游系统解析；链式调用从左到右执行，顺序不影响结果（两个关键词互不包含）。

### 常见错误

| 错误现象 | 原因 | 修正 |
|---|---|---|
| 字段仍带空格 | 只清洗了整行 | 逐字段 `strip()` |
| 字段数为 3 | 用了 `split()` 无参（按空白切） | 写 `split("\|")` |
| 校验为 `False` | 用 `len(parts)` 与 `count` 直接比较 | 应为 `count == len(parts) - 1` |
| `IndexError` | 循环写成 `range(len(parts) + 1)` | 上界不 `+1` |

## 答案 4-2　路段名称检索与脱敏

> 对应 **练习 4-2**（第三部分）

### 参考代码

````python
# ---------- 第 04 章 字符串与输入输出 · 练习 4-2 参考答案：路段名称检索与脱敏 ----------
# 对应知识点：4.3 in / find() / startswith() / endswith() / count()、4.2 replace()
# 功能说明：在路段名中检索敏感关键词，统计出现次数与位置，并输出脱敏后的名称

name = input("路段名称：")
keyword = input("敏感关键词：")

print("是否包含关键词：", keyword in name)
print("首次出现位置：", name.find(keyword))
print("是否以'京'开头：", name.startswith("京"))
print("是否以'高速'结尾：", name.endswith("高速"))
print("关键词出现次数：", name.count(keyword))

masked = name.replace(keyword, "***")
print("脱敏后名称：", masked)

if name.find(keyword) == -1:      # find() 找不到时返回 -1，而不是抛异常
    print("结论：无需脱敏")
else:
    print("结论：已脱敏，可对外发布")
````

```text
路段名称：京港澳高速
敏感关键词：京
是否包含关键词： True
首次出现位置： 0
是否以'京'开头： True
是否以'高速'结尾： True
关键词出现次数： 1
脱敏后名称： ***港澳高速
结论：已脱敏，可对外发布
```

### 解析

1. **`in` 与 `find` 的分工**：`in` 只回答"有没有"，`find` 回答"在哪"。本例两者都需要，故各用一次；若只需布尔结论，用 `in` 即可，不必比较 `-1`。
2. **位置为 0 的陷阱**：若写成 `if name.find(keyword):` 判断，位置 0 会被当成假值，导致"无需脱敏"的错误结论。必须写 `== -1` 或 `!= -1`。
3. **`count` 与 `replace` 的一致性**：`count` 统计 1 次，`replace` 也只替换 1 处，二者同源（都按子串完全匹配）。
4. **脱敏的业务含义**：把敏感词替换为 `***` 后仍保留长度信息，既能对外发布又不泄露具体名称。

### 常见错误

| 错误现象 | 原因 | 修正 |
|---|---|---|
| 位置 0 被判为"无需脱敏" | `if name.find(k):` | `== -1` |
| 脱敏结果为空 | 用 `replace(keyword, "")` | 替换为 `"***"` |
| `startswith` 大小写不符 | 名称含小写 | 先统一 `upper()` |
| 用 `find` 与 `in` 重复判断 | 代码冗余 | 择一即可 |

## 本章小结

- **索引**：从 0 开始，负索引从末尾数；**切片**左闭右开，越界切片不报错。
- **清洗**：`strip` 去两端空白、`upper` / `lower` 统一大小写、`replace` 替换子串；字符串**不可变**，必须接收返回值。
- **检索**：`in` 判有无、`find` 给位置（找不到 `-1`）、`startswith` / `endswith` 判前后缀、`count` 数次数。
- **切分**：`split` 切、`join` 拼；段数 = 分隔符数 + 1，可用于数据校验。
- **格式化**：`f"{值:说明符}"` 是主流；`:<` `:>` `:^` 控对齐，`:.1f` 控精度，宽度须表头数据一致。
