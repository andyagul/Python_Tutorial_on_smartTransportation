## 3.2 知识点 3.1　`if` / `elif` / `else` 与缩进

### 3.2.1 核心概念

`if` 后面跟一个**条件表达式**（结果为 `True` / `False`），条件成立则执行缩进的代码块：

- **冒号 + 缩进**是 Python 判断语句的固定格式，缩进通常 4 个空格；
- 同一缩进层级的语句属于同一个代码块，缩进结束即代码块结束；
- `elif`（else if 的缩写）用于多分支，可写多个；
- `else` 兜底，不需要条件，最多一个。

**原理补充（概念深化）**：Python 用**缩进**而不是大括号界定代码块，这在编译阶段就决定了语句归属：解释器靠 INDENT / DEDENT 记号判断「哪些行属于这个分支」，因此同一个块里的缩进量必须一致，混用 Tab 与空格会直接报错。多分支 `if / elif / else` 的关键机制是**自上而下短路**：一旦某个条件成立，执行完它的块就跳过后面所有 `elif` 与 `else`，哪怕后面的条件也为真。这意味着分支顺序本身就是业务规则——把宽条件写在窄条件前面，窄条件永远命中不了，正是第 5A 节反复训练的「不报错、只算错」bug。写分级判断时应当按阈值从高到低（或从严到宽）排列，并用一个边界值逐个验一遍。`else` 不带条件、最多一个，它的价值是兜住所有没被前面覆盖的情况，避免变量未定义。

### 3.2.2 语法要点

| 结构 | 适用 | 示例 |
|---|---|---|
| `if` | 单一条件 | `if speed < 20:` |
| `if … else` | 二选一 | `if queue > 0: … else: …` |
| `if … elif … else` | 多区间互斥 | 四级拥堵判定 |
| `if` 嵌套 | 条件有依赖关系 | 先判是否超速，再判是否货车 |

**多分支的两条铁律**：

1. **自上而下匹配，命中即跳出**：一旦某个分支成立，其后的 `elif` / `else` 全部跳过；
2. **区间必须"从严格到宽松"排列**：若先写 `speed >= 20`，后面的 `speed >= 60` 永远不会执行。

### 3.2.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 忘记冒号 | `SyntaxError: expected ':'` | 每个 `if` / `elif` / `else` 后加 `:` |
| 缩进不一致 | `IndentationError` | 全文件统一 4 空格，不用 Tab 混排 |
| 顺序写反 | 高级别分支永不触发 | 严格条件在前 |
| 用 `=` 判断 | `SyntaxError` | `==` |
| `else` 后写条件 | `SyntaxError` | `else` 不带条件，需要条件用 `elif` |

### 3.2.4 随堂示例：四档拥堵分级

**核心要点**：验证 `if-elif-else` 自顶向下逐条判断、命中即跳出：同一份数据只会落进一个分支，`else` 兜住所有剩余情况。

````python
speed = 35

if speed >= 60:
    level = "畅通"
elif speed >= 40:
    level = "缓行"
elif speed >= 20:
    level = "拥堵"
else:
    level = "严重拥堵"

print("车速", speed, "km/h →", level)
print("提示：elif 的顺序决定结果，改动顺序会改变分级")
````

**预期输出**：

```text
车速 35 km/h → 拥堵
提示：elif 的顺序决定结果，改动顺序会改变分级
```

**变式代码（迁移练习）**：把随堂示例的单次分级迁移为「同一套分支跑两个不同车速」，直观看到条件顺序如何决定走哪条路径。

````python
# 多分支自上而下判断，命中第一个成立条件后立即跳出整条链
speed = 45
if speed >= 60:
    level = "畅通"
elif speed >= 40:
    level = "缓行"
else:
    level = "拥堵"
print("车速 45 →", level)
# 同一套分支换个输入，会走完全不同的路径
speed = 72
if speed >= 60:
    level = "畅通"
elif speed >= 40:
    level = "缓行"
else:
    level = "拥堵"
print("车速 72 →", level)
````

**预期输出**：

```text
车速 45 → 缓行
车速 72 → 畅通
```

### 3.2.5 实操步骤（上机跟随）

1. 先按代码顺序手推两次判定的走法，再运行，确认输出是「缓行 / 畅通」。
2. 把第一个条件改成 `speed >= 40`，观察 45 也被判成「畅通」，理解宽条件在前会吃掉窄条件。
3. 把 `speed = 45` 改成恰好 `40`，验证 `>=` 的边界归入哪一级。
4. 删掉最后的 `else`，并把输入改成 20，观察 `level` 变成未定义，理解兜底分支的必要性。
5. 把某个分支的缩进从 4 空格改成 2 空格，记录 `IndentationError`，再改回。

### 3.2.6 常见问题 FAQ

| 现象 | 根因 | 处理 |
|---|---|---|
| `IndentationError` | 同一代码块缩进不一致或混用了 Tab 与空格 | 统一 4 个空格，编辑器开启「显示空白字符」 |
| 所有输入都判成第一级 | 把宽条件写在了前面 | 按阈值从高到低重排，或用 `40 <= speed < 60` 显式限定区间 |
| `SyntaxError: expected ':'` | 条件后漏写冒号 | `if 条件:` 末尾必须有英文冒号 |
| 某个分支下变量没定义 | 并非所有路径都给变量赋了值 | 用 `else` 兜底，或在 `if` 前先给一个默认值 |

## 3.3 知识点 3.2　条件表达式（三元写法）

### 3.3.1 核心概念

当"二选一"的赋值只有一行时，可以写成 `值A if 条件 else 值B`：条件为真取值 A，否则取值 B。它不是新功能，只是 `if-else` 的紧凑写法。

```text
state = "拥堵" if p > 0.6 else "不拥堵"
```

**原理补充（概念深化）**：`值A if 条件 else 值B` 是一个**表达式**而不是语句：它先求条件，再只求被选中的那一支，最后把结果作为整体值交出去，因此可以直接出现在赋值右侧、函数实参甚至 `print()` 里。这与 `if-elif-else` 语句的本质差别是：语句决定「执行哪段代码」，表达式决定「算出哪个值」。也正因为只求选中分支，右侧放一个有副作用的动作时要格外小心。三元写法适合**二选一且两支都很短**的场合；一旦嵌套到两层以上，可读性会迅速下降，应改回 `if` 或查表法。常见误用是把 `if` 条件写得很长、两支各含多行逻辑，那已经不是「紧凑」而是「压缩」。记住一条判断标准：如果这行需要读者回头看第二遍才能懂，就拆开写。

### 3.3.2 语法要点

| 写法 | 等价形式 |
|---|---|
| `x = a if cond else b` | `if cond: x = a` / `else: x = b` |
| `print("是" if flag else "否")` | 可直接嵌在函数参数中 |

### 3.3.3 常见用法与易错点

- **只适合"二选一"**：三个以上分支仍应写 `elif` 链，否则可读性急剧下降。
- 顺序是"值—条件—值"，不要写成 `if cond: a else b`（那是别的语言的写法）。
- 不要嵌套多层三元表达式，这是可读性杀手。

### 3.3.4 随堂示例：用三元写法一行定等级

**核心要点**：验证条件表达式 `A if 条件 else B` 的求值方式：条件为真取前者，为假取后者；适合「二选一」的赋值，不适合三个以上分支。

````python
speed = 72

level = "畅通" if speed >= 60 else "非畅通"
print("等级：", level)

# 三元写法也可以直接嵌在计算里
green = 30 if level == "畅通" else 50
print("绿灯时长：", green, "秒")

# 两个分支都要求值一次，因此适合简单判断
print("是否超速：", "是" if speed > 120 else "否")
````

**预期输出**：

```text
等级： 畅通
绿灯时长： 30 秒
是否超速： 否
```

**变式代码（迁移练习）**：把随堂示例的「二选一赋值」迁移到「直接嵌入计算 + 两层嵌套三选一」，对比简洁性与可读性的临界点。

````python
# 三元写法适合「二选一赋值」，右侧先算条件再选值
speed = 52
level = "畅通" if speed >= 60 else "缓行"
print("等级：", level)
# 也可以直接嵌在计算里，省掉中间变量
green = 30 if speed >= 60 else 50
print("绿灯：", green, "秒")
# 嵌套三元能表达三选一，但超过两层应改回 if
level2 = "畅通" if speed >= 60 else ("缓行" if speed >= 40 else "拥堵")
print("三选一：", level2)
````

**预期输出**：

```text
等级： 缓行
绿灯： 50 秒
三选一： 缓行
```

### 3.3.5 实操步骤（上机跟随）

1. 预测三行输出（缓行 / 50 秒 / 缓行）后再运行核对。
2. 把 `speed` 改成 72，观察 `green` 变成 30，确认三元表达式随输入实时求值。
3. 把嵌套三元的阈值 `40` 改成 `60`，观察 52 被判成「拥堵」，理解条件顺序同样有效。
4. 把 `level2` 那行改写成等价的 `if/elif/else`，比较两种写法的行数与可读性。
5. 试着再套一层三元，读完一遍后判断：这行是否应该拆开？把结论写成注释。

### 3.3.6 常见问题 FAQ

| 现象 | 根因 | 处理 |
|---|---|---|
| 把 `if` 写在最前面报语法错 | 三元表达式的顺序是 `A if 条件 else B` | 记住「值在前、条件居中、备选取后」 |
| 嵌套三元看不懂 | 两层以上分支让单行信息量过载 | 改回 `if/elif/else` 或用字典查表 |
| 两个分支都被执行了 | 误以为表达式会两边都算 | 只求被选中的一支；必须执行的动作放到表达式外 |
| 忘写 `else` 报语法错 | 三元表达式必须给出两个候选值 | 补 `else 备选值`，哪怕只是一个空串 |

## 3.4 知识点 3.3　`for` 与 `range()`

### 3.4.1 核心概念

`for` 用于**已知次数**的重复：遍历一个序列，每次取出一个值执行循环体。本章的序列由 `range()` 生成等差整数列。

`range()` 三种写法：

| 写法 | 生成 | 元素个数 |
|---|---|---|
| `range(5)` | 0,1,2,3,4 | 5 |
| `range(6, 22, 2)` | 6,8,10,…,20 | 8 |
| `range(1, 5)` | 1,2,3,4 | 4 |

**含头不含尾**：终点值本身不在序列中，这是 `range` 最容易出错的地方。

**原理补充（概念深化）**：`range(start, stop, step)` 返回的是一个**惰性序列对象**：它不预先生成所有整数，只记住三个参数，每次被取用时算下一个值，所以 `range(1, 1000000)` 几乎不占内存。规则是「含头不含尾」，`range(6, 13, 2)` 产生 6、8、10、12 共 4 个值；`step` 为负时 `start` 必须大于 `stop`，否则一个值都不产生。`for` 的本质是**遍历可迭代对象**：每轮把下一个值绑定到循环变量，执行完块内语句再回来取下一个，循环变量在结束后仍保留最后一轮的值。理解了「含头不含尾」，`len(range(...))` 就能直接当循环次数用，不必再手工数。统计类程序里最常见的两类错误是：把 `range` 的 `stop` 写成实际末值（少算一轮），以及在循环体内重复初始化累加器（结果只剩最后一轮）。

### 3.4.2 语法要点

- 循环变量（如 `hour`）由 `for` 自动赋值，无需事先定义；
- 循环体必须缩进；
- 循环变量的值在循环结束后仍保留（最后一次的值）；
- 需要"第几次"时，用 `range(1, n + 1)` 让变量从 1 开始，比在循环内 `i + 1` 直观。

### 3.4.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 漏掉终点 | `range(1, 8)` 只得 7 个数 | 需要 1—8 写 `range(1, 9)` |
| 步长写成 0 | `ValueError: range() arg 3 must not be zero` | 步长为负可倒序，不能为 0 |
| 循环体内改了循环变量 | 下一次迭代被覆盖，改动无效 | 需要跳过用 `continue` |
| 循环外忘记缩进 | 语句只执行一次 | 检查缩进层级 |

### 3.4.4 随堂示例：range() 的三个参数

**核心要点**：验证 `range(start, stop, step)` 的左闭右开特性（不含 `stop`）、步长作用，以及用循环变量参与计算得到逐行结果。

````python
total = 0
for hour in range(6, 13, 2):        # 6,8,10,12：不含 13
    flow = hour * 100               # 用循环变量算出一个"车流量"
    total += flow
    print(hour, "时 车流", flow, "辆")

print("合计：", total, "辆")
print("循环次数：", len(range(6, 13, 2)))
````

**预期输出**：

```text
6 时 车流 600 辆
8 时 车流 800 辆
10 时 车流 1000 辆
12 时 车流 1200 辆
合计： 3600 辆
循环次数： 4
```

**变式代码（迁移练习）**：把随堂示例的「步长遍历累加」迁移为「正序累加 + 负序倒序遍历」同框对照，验证步长符号决定方向。

````python
# range(start, stop, step) 含头不含尾，step 可为负
total = 0
for h in range(6, 13, 2):
    total = total + h * 100
    print(h, "时 车流", h * 100, "辆")
print("合计：", total, "辆")
print("循环次数：", len(range(6, 13, 2)))
# 步长写成负数就是倒序，start 必须大于 stop
for h in range(12, 5, -2):
    print("倒序时段：", h)
````

**预期输出**：

```text
6 时 车流 600 辆
8 时 车流 800 辆
10 时 车流 1000 辆
12 时 车流 1200 辆
合计： 3600 辆
循环次数： 4
倒序时段： 12
倒序时段： 10
倒序时段： 8
倒序时段： 6
```

### 3.4.5 实操步骤（上机跟随）

1. 先列出 `range(6, 13, 2)` 的四个值，再运行核对输出的车流合计 3600。
2. 把 `13` 改成 `12`，观察少一轮（12 时被排除），验证「含头不含尾」。
3. 核对「循环次数」是否等于实际打印的时段行数，理解 `len(range(...))` 的用途。
4. 把 `-2` 改成 `2`，观察倒序循环一次都不执行，理解负步长对 `start > stop` 的要求。
5. 把 `total = 0` 移到循环体内，观察合计只剩 1200，复现 05A 节练过的累加器位置 bug。

### 3.4.6 常见问题 FAQ

| 现象 | 根因 | 处理 |
|---|---|---|
| 循环少跑一轮 | 把 `stop` 当成了包含的末值 | 写成 `末值 + 1`，或用 `len()` 核对次数 |
| 倒序循环没有任何输出 | 负步长时 `start` 不大于 `stop` | 改成 `range(12, 5, -2)` 这类合法区间 |
| `range(0, 5, 0)` 报错 | 步长不能为 0 | 步长必须是非零整数 |
| 循环结束后变量还能用 | 循环变量在 `for` 结束后仍绑定最后一轮的值 | 不要依赖它做后续判断，需要就显式另存 |

## 3.5 知识点 3.4　`while`、`break`、`continue` 与循环 `else`

### 3.5.1 核心概念

`while` 用于**次数未知、由条件决定**的重复：只要条件为真就继续。典型场景是"排队车辆未清空就继续放行"。

三个控制语句：

| 语句 | 作用 | 注意 |
|---|---|---|
| `break` | 立即跳出整个循环 | 其后的循环体语句不再执行 |
| `continue` | 跳过本轮剩余语句，进入下一轮 | 不结束循环 |
| `else`（挂在循环上） | 循环**未被 `break` 中断**时执行 | 与 `if-else` 的 `else` 语义完全不同 |

`while … else …` 的价值正在于区分两种结局："正常跑完"与"被中断"。

**原理补充（概念深化）**：`while` 把「是否继续」交给条件表达式，每轮开始前重新求值一次，因此**循环体内必须有让条件趋近为假的更新语句**，否则就是死循环。`break` 立即终止整个循环，`continue` 只跳过本轮剩余语句、随后照常回到条件判断——两者作用范围完全不同，混淆时常见表现是「该停的没停」或「少算一轮」。工程上给 `while` 加一个轮次上限作为安全阀是必要的，因为业务数据一旦异常，条件可能永远为真。循环 `else` 则只在**循环正常耗尽条件、没有触发 `break`** 时执行，它更像「未被中断」标记而不像 `if` 的否则分支。与 `for` 的选择标准很直接：次数已知用 `for`，次数由数据状态决定用 `while`。

### 3.5.2 语法要点

```text
while 条件:
    循环体
else:
    正常结束时执行
```

- `while` 的条件必须在循环体内被改变，否则形成死循环；
- `continue` 在 `while` 中若跳过了条件变量的更新语句，会造成死循环——这是最常见的 `continue` 事故；
- `for … else` 也合法，语义同上。

### 3.5.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 忘记更新条件变量 | 程序卡死、无限输出 | 循环体内必须有 `queue -= per_min` 之类 |
| `continue` 前忘了 `minute += 1` | 死循环 | 把"计时/计数"放在 `continue` 之前 |
| 以为 `else` 总会执行 | `break` 后 `else` 不执行 | 记住"未被中断才执行" |
| 用 `break` 代替条件 | 逻辑难读 | 能用条件表达就不用 `break` |

### 3.5.4 随堂示例：while + break + 循环 else

**核心要点**：验证三件事：`while` 由条件控制次数、`break` 立即跳出整个循环、循环 `else` 只在「未被 `break` 中断」时执行。

````python
def dispatch(queue, limit, per_min):
    """每分钟放行 per_min 辆；超过 limit 分钟则转人工"""
    minute = 0
    while queue > 0:                 # 条件为真才继续循环
        minute += 1
        queue = max(queue - per_min, 0)      # 不会减成负数
        print("  第", minute, "分钟，剩余", queue, "辆")
        if minute >= limit:
            print("  等待超时，转人工疏导")
            break                    # 立即跳出整个循环
    else:                            # 只有「未被 break 中断」才执行
        print("  排队已清空，共耗时", minute, "分钟")

print("【情形一】排队 6 辆，限 3 分钟：")
dispatch(6, 3, 3)                    # 2 分钟清空 → 走 else

print("【情形二】排队 10 辆，限 3 分钟：")
dispatch(10, 3, 3)                   # 3 分钟未清空 → 走 break
````

**预期输出**：

```text
【情形一】排队 6 辆，限 3 分钟：
  第 1 分钟，剩余 3 辆
  第 2 分钟，剩余 0 辆
  排队已清空，共耗时 2 分钟
【情形二】排队 10 辆，限 3 分钟：
  第 1 分钟，剩余 7 辆
  第 2 分钟，剩余 4 辆
  第 3 分钟，剩余 1 辆
  等待超时，转人工疏导
```

**变式代码（迁移练习）**：把随堂示例的排队放行迁移为「`continue` 跳过打印 + `break` 安全阀」同框演示，看清两者作用域差别。

````python
# while 用于次数未知：队列未清空就继续；break 提供紧急出口
queue = 23
minute = 0
while queue > 0:
    minute = minute + 1
    queue = queue - 6
    if queue < 0:
        queue = 0
    if minute > 10:        # 安全阀：超时转人工
        print("超时，转人工")
        break
    if minute == 3:
        continue           # 跳过本轮打印，循环条件照常判断
    print("第", minute, "分钟，剩余", queue, "辆")
print("结束于第", minute, "分钟")
````


### 3.5.6 常见问题 FAQ

| 现象 | 根因 | 处理 |
|---|---|---|
| 程序卡住不动 | 循环体内没有让条件变假的更新语句 | 检查每轮是否推进状态；先加轮次上限再排查 |
| `continue` 之后少算了一轮数据 | 把累加写在了 `continue` 之后 | 把必须执行的更新移到 `continue` 之前 |
| `break` 之后 `else` 块没执行 | 循环 `else` 只在未被中断时执行 | 需要无论是否中断都执行就放 `finally` 或移到循环外 |
| 边界数据多跑了一轮 | 条件用 `>= 0` 而队列已被夹到 0 | 把条件改成 `> 0`，配合 `max(..., 0)` 夹值 |

## 3.6 知识点 3.5　循环三要素与嵌套循环

### 3.6.1 核心概念

统计类程序的骨架是固定的，称为**循环三要素**：

1. **初始化**：在循环**外**设置初值（`total = 0`、`peak = 0`）；
2. **更新**：在循环**内**累加、计数或比较（`total += flow`）；
3. **输出**：循环**结束后**统一输出（`print(total)`）。

三者缺一不可，最常见的错误是把初始化写进循环——那样每轮都清零，结果永远是最后一轮的值。

**嵌套循环**是"二维"的重复：外层控制行（车道），内层控制列（车位）。执行次数为外层次数 × 内层次数。

**原理补充（概念深化）**：统计类程序有固定骨架：**循环外初始化、循环内更新、循环后输出**。三者的位置由作用域决定：初始化若放进循环体，每轮都被重置，结果只剩最后一轮；输出若放进循环体，得到的是逐轮中间值而不是最终报表。嵌套循环则是「外层每走一轮，内层完整跑一圈」，总执行次数是两层次数之积，因此三层以上嵌套要警惕耗时。内层若需要「每行重新开始」的变量（如拼一行的字符串），初始化必须放在**外层循环体内、内层循环之前**——这是最容易放错的位置。判断放得对不对有个简单办法：问自己「这个值应该活多久」，活整段程序就放最外，活一行就放外层循环内，活一格就放内层循环内。第 5A 节的断点练习里，`total` 每轮归零就是这条规则被破坏的典型。

### 3.6.2 语法要点

| 模式 | 写法 | 交通场景 |
|---|---|---|
| 累加求和 | `total += x` | 全天总流量 |
| 计数 | `count += 1` | 超速车辆数 |
| 求极值 | `if x > peak: peak = x` | 峰值时段 |
| 极值 + 位置 | `if x > peak: peak = x; peak_hour = hour` | 峰值出现在哪个时段 |
| 二维遍历 | 外层 `for lane` 内层 `for slot` | 车道 × 车位网格 |

### 3.6.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 初始化放在循环内 | 累加结果只等于最后一个值 | 移到循环外 |
| 极值初值设 0 | 全为负数时结果错误 | 用第一个元素初始化，或设足够小的初值 |
| 极值与位置不同步 | 峰值对了但时段错了 | 两个变量在同一个 `if` 内更新 |
| 内层 `print` 未拼接 | 每个元素占一行 | 先拼字符串，内层结束再整行输出 |
| 内外层变量重名 | 逻辑混乱 | 外层 `lane`、内层 `slot`，命名区分 |

### 3.6.4 随堂示例：嵌套循环打印车道占用网格

**核心要点**：验证「外层管行、内层管列」的嵌套结构，并强调两个易错点：每行开始要重置拼接变量，整行拼完再输出一次。

````python
lanes = 3
slots = 4
occupied = 2                 # 每车道前 2 个车位已占用

for lane in range(1, lanes + 1):        # 外层：逐条车道
    line = ""                           # 每行开始必须重置
    for slot in range(1, slots + 1):    # 内层：逐个车位
        if slot <= occupied:
            line += "[X]"
        else:
            line += "[ ]"
    print("第", lane, "车道：", line)    # 整行拼完再输出
````

**预期输出**：

```text
第 1 车道： [X][X][ ][ ]
第 2 车道： [X][X][ ][ ]
第 3 车道： [X][X][ ][ ]
```

**变式代码（迁移练习）**：把随堂示例的车位网格迁移为「双层循环求路网合计与峰值」，同时演示三个变量各自应放在哪一层初始化。

````python
# 三要素：循环外初始化 → 循环内更新 → 循环后输出
total = 0
peak = 0
for lane in range(1, 4):              # 外层：3 条车道
    for slot in range(1, 4):          # 内层：每条车道 3 个时段
        flow = lane * 100 + slot * 10
        total = total + flow
        if flow > peak:
            peak = flow
    print("车道", lane, "累计", total, "辆")
print("全路网合计：", total, "辆 | 峰值：", peak, "辆")
````

**预期输出**：

```text
车道 1 累计 360 辆
车道 2 累计 1020 辆
车道 3 累计 1980 辆
全路网合计： 1980 辆 | 峰值： 330 辆
```

### 3.6.5 实操步骤（上机跟随）

1. 手算每条车道的三个流量，预测三次「累计」与最终合计 1980，再运行核对。
2. 确认 `peak` 最终是 330，说明它在内层逐格比较、且初始化在最外层。
3. 把 `total = 0` 移进外层循环体内，观察每条车道都从 0 重新累加，报表失真。
4. 把最后的 `print` 缩进到外层循环里，观察「全路网合计」被打印三次且都是中间值。
5. 把内层 `range(1, 4)` 改成 `range(1, 5)`，预测总执行次数从 9 变 12，再运行验证。

### 3.6.6 常见问题 FAQ

| 现象 | 根因 | 处理 |
|---|---|---|
| 合计只剩最后一轮的值 | 初始化写进了循环体 | 把 `total = 0` 移到所有循环之外 |
| 每行拼接内容越来越长 | 「每行重置」的变量放到了最外层 | 放在外层循环体内、内层循环之前 |
| 嵌套循环跑得极慢 | 执行次数是各层次数之积 | 减少层数，或把能提前算的部分移到内层之外 |
| 内外层用了同一个循环变量 | 内层覆盖了外层的值，导致外层逻辑错乱 | 用语义化且互不重名的变量，如 `lane` 与 `slot` |

- **源文件**：`case2_分时段车流统计.py`

### 任务情境

调查员需要按 2 小时一个时段采集全天 8 个时段的车流量，统计总量与平均流量，找出峰值时段，并对超过阈值的时段给出告警。

### 任务要求

1. 用 `for` 与 `range(6, 22, 2)` 生成 6、8、…、20 共 8 个时段；
2. 循环内读取流量并累加总量，同时用 `if` 维护峰值与峰值时段；
3. 单时段流量超过 1500 辆时输出告警；
4. 循环结束后输出总量、平均流量、峰值时段。

### 参考代码

````python
# ---------- 第 03 章 流程控制 · 案例 3-2：全天分时段车流统计 ----------
# 对应知识点：3.3 for 与 range() 的步长、3.5 累加器与极值维护、3.1 循环体内的分支
# 功能说明：采集 8 个时段的车流量，统计总量、峰值与峰值时段，并对超 1500 辆的时段告警

# 全天分时段车流采集与统计
total = 0        # 全天总量
peak = 0         # 峰值流量
peak_hour = 0    # 峰值所在时段

for hour in range(6, 22, 2):          # 6,8,10,12,14,16,18,20
    flow = eval(input("第 " + str(hour) + " 时段车流量(辆/2h)："))
    total += flow
    if flow > peak:
        peak = flow
        peak_hour = hour
    if flow > 1500:
        print("  [告警] " + str(hour) + " 时段流量偏高：" + str(flow) + " 辆")

print("全天总流量：", total, "辆")
print("平均每时段：", round(total / 8, 1), "辆")
print("峰值时段：", peak_hour, "时，流量", peak, "辆")
print("峰值是否超过 1500：", peak > 1500)
````

# 第四部分　习题答案

## 答案 3-1　车辆排队放行仿真

> 对应 **练习 3-1**（第三部分）

### 参考代码

````python
# ---------- 第 03 章 流程控制 · 练习 3-1 参考答案：车辆排队放行仿真 ----------
# 对应知识点：3.4 while 循环、break 跳出、循环 else 子句
# 功能说明：模拟排队车辆按每分钟放行数逐渐清空；等待超过 30 分钟则转人工疏导

queue = eval(input("排队车辆数："))
per_min = eval(input("每分钟放行车辆数："))
minute = 0

while queue > 0:              # 队列未清空就一直放行
    minute += 1               # 计数写在循环体首行，避免从 0 分钟开始
    if minute > 30:
        print("等待超过 30 分钟，转人工疏导")
        break
    queue -= per_min          # 本分钟放行后剩余排队
    if queue < 0:
        queue = 0             # 防止出现负数排队
    print("第 " + str(minute) + " 分钟，剩余排队", queue, "辆")
else:
    print("排队已清空，共耗时", minute, "分钟")
````

### 运行结果（示例输入：147 / 23）

```text
排队车辆数：147
每分钟放行车辆数：23
第 1 分钟，剩余排队 124 辆
第 2 分钟，剩余排队 101 辆
第 3 分钟，剩余排队 78 辆
第 4 分钟，剩余排队 55 辆
第 5 分钟，剩余排队 32 辆
第 6 分钟，剩余排队 9 辆
第 7 分钟，剩余排队 0 辆
排队已清空，共耗时 7 分钟
```

### 解析

1. **`minute` 的位置**：先 `minute += 1` 再判断超时，保证"第 1 分钟"从 1 开始而不是 0。
2. **`break` 与 `else` 的配合**：超时走 `break`，`else` 不执行，因此**不会**误报"共耗时"；正常清空时循环条件自然为假，`else` 执行。这正是知识点 3.4 的用途。
3. **负数归零**：第 7 轮 `9 - 23` 得 `-14`，若不归零会输出"剩余排队 -14 辆"。
4. **为什么用 `while` 而不是 `for`**：放行轮数由排队数与放行速率决定，事先未知，符合 `while` 的适用条件。

### 常见错误

| 错误现象 | 原因 | 修正 |
|---|---|---|
| 程序卡死 | 忘记 `queue -= per_min` | 循环体内更新条件变量 |
| 输出"剩余 -14 辆" | 未做负数归零 | 减完后 `if queue < 0: queue = 0` |
| 超时后仍输出"共耗时" | 用了 `if` 而非 `break` | 用 `break` 跳出，让 `else` 不执行 |
| 从"第 0 分钟"开始 | `minute += 1` 写在输出之后 | 移到循环体首行 |

| `while` 结构 | 3 | 条件正确、有更新语句 |
| `break` 超时分支 | 3 | 30 分钟后跳出且不输出总用时 |
| 循环 `else` | 2 | 正常清空时输出总用时 |
| 边界处理 | 2 | 负数归零、分钟数从 1 开始 |

## 答案 3-2　打印路口车道占用网格

> 对应 **练习 3-2**（第三部分）

### 参考代码

````python
# ---------- 第 03 章 流程控制 · 练习 3-2 参考答案：打印路口车道占用网格 ----------
# 对应知识点：3.5 循环三要素与嵌套循环、3.1 内层分支、字符串累加拼接
# 功能说明：按车道数与每车道车位数打印占用网格：[X] 表示已占用、[ ] 表示空位

lanes = eval(input("车道数："))
slots = eval(input("每车道车位数："))
occupied = eval(input("每车道已停车数："))

for lane in range(1, lanes + 1):        # 外层：逐条车道
    line = ""                           # 每换一条车道必须重置，否则上一行会累积进来
    for slot in range(1, slots + 1):     # 内层：逐个车位拼接
        if slot <= occupied:
            line = line + "[X]"         # 已占用
        else:
            line = line + "[ ]"         # 空位
    print("第 " + str(lane) + " 车道：" + line)   # 内层结束后整行输出一次
````

### 运行结果（示例输入：4 / 8 / 3）

```text
车道数：4
每车道车位数：8
每车道已停车数：3
第 1 车道：[X][X][X][ ][ ][ ][ ][ ]
第 2 车道：[X][X][X][ ][ ][ ][ ][ ]
第 3 车道：[X][X][X][ ][ ][ ][ ][ ]
第 4 车道：[X][X][X][ ][ ][ ][ ][ ]
```

### 解析

1. **`line = ""` 的位置**：必须在外层循环内、内层循环前，这样每条车道都从空串开始拼接；若写到外层循环之外，第二条车道会带着第一条的内容。
2. **为什么先拼后打**：`print` 默认换行。若在内层 `print("[X]")`，每个车位占一行；拼接成 `line` 后外层只打一次，才得到"一行一条车道"。
3. **`range(1, lanes + 1)`**：车道编号从 1 开始，终点 `+1` 才包含第 4 条车道（含头不含尾）。
4. **`slot <= occupied`**：已停车数为 3 时，第 1—3 个车位为 `[X]`，第 4—8 个为空，符合"从前向后占用"的业务假设。

### 常见错误

| 错误现象 | 原因 | 修正 |
|---|---|---|
| 只有一行网格 | `line` 写在两层循环之外 | 移入外层循环体首行 |
| 网格逐格换行 | 内层直接 `print` | 拼接后整行输出 |
| 只有 3 条车道 | `range(1, lanes)` | 改为 `range(1, lanes + 1)` |
| 网格累积到下一行 | `line` 未重置 | 每轮外层开始时置空 |

### 评分要点（供批改）

| 项 | 分值 | 观察点 |
|---|---|---|
| 嵌套结构 | 4 | 外层车道、内层车位，层次正确 |
| 拼接输出 | 3 | 一行一条车道，无错行 |
| 占用判断 | 2 | `[X]` / `[ ]` 数量正确 |
| 编号 | 1 | 车道从 1 开始 |

## 本章小结

- **分支**：`if` → `elif` → `else` 自上而下匹配、命中即跳出；区间**严格在前**；`else` 不带条件。
- **三元**：`a if cond else b` 只用于二选一。
- **`for`**：已知次数用 `for` + `range()`，记住**含头不含尾**。
- **`while`**：次数未知用 `while`，条件变量必须在循环体内更新。
- **控制**：`break` 跳出、`continue` 跳过、循环 `else` 仅在未被 `break` 时执行。
- **三要素**：初始化在外、更新在内、输出在后；极值与位置必须同步更新。
- **嵌套**：外层行、内层列；先拼接再整行输出。
