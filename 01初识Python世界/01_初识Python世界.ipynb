## 1.2 知识点 1.1　输出语句与字符串字面量

### 1.2.1 核心概念

程序与外界交流的三种方式中，最基础的是**输出**（另外两种是输入与文件读写，分别在第 2 章、第 8 章）。Python 用 `print()` 完成输出：

- `print` 是**函数**，后面跟一对圆括号，括号内写要输出的内容；
- 括号内写的东西叫**参数**，本章只有一个参数：一段**字符串**；
- 字符串必须用引号括起来，引号内的内容原样输出（不含引号本身）；
- 每次调用 `print()` 输出完内容后**自动换行**，所以 5 条 `print` 就是 5 行。

### 1.2.2 语法要点

| 语法形式 | 说明 | 输出 |
|---|---|---|
| `print("文本")` | 双引号字符串 | 文本 |
| `print('文本')` | 单引号字符串，与双引号等价 | 文本 |
| `print("")` | 空字符串，输出一个空行 | 空行 |
| `print("a", "b")` | 多参数用逗号分隔，中间自动补一个空格 | `a b` |

**引号的三种选择**：

1. 文本本身不含引号：单引号、双引号随意，推荐统一用双引号；
2. 文本内含双引号：外层用单引号，如 `print('屏幕显示"限速 60"')`；
3. 文本内含单引号：外层用双引号，如 `print("Don't stop")`。

### 1.2.3 常见用法与易错点

| 情况 | 错误写法 | 后果 | 正确写法 |
|---|---|---|---|
| 漏写引号 | `print(智慧交通)` | `NameError: name '智慧交通' is not defined` | `print("智慧交通")` |
| 引号不成对 | `print("提示)` | `SyntaxError`，后续整段都无法运行 | 成对书写，编辑器会高亮配对 |
| 中文标点 | `print（"提示"）` | `SyntaxError: invalid character` | 括号、引号、逗号一律用**英文半角** |
| 用中文引号 | `print(“提示”)` | `SyntaxError` | 用 `"` 或 `'` |
| 误以为引号会输出 | —— | 输出中不含引号，属正常现象 | 需要引号本身时用 `\"`，见 1.3 |

> **课堂提示**：`SyntaxError`（语法错误）在程序**运行前**就会被发现，报错信息会指出行号；而 `NameError` 是运行到该行才出现。初学者九成的报错来自中文标点与引号不成对，先查这两项。

### 1.2.4 随堂示例：四种 print 写法一次跑通

**核心要点**：验证 `print()` 的四种基本写法：单引号、双引号、多参数（自动补空格）、空参数（输出空行），并确认引号本身不会出现在输出里。

````python
# 单引号与双引号等价，引号本身不会输出
print('限速 60')
print("车道 2 畅通")

# 多参数用逗号分隔，输出时自动补一个空格
print("京A12345", "62", "km/h")

# 空参数：只输出一个空行
print()

# 空字符串：同样输出一个空行
print("")
print("（上面第 4、5 条 print 各产生一个空行）")
````

**预期输出**：

```text
限速 60
车道 2 畅通
京A12345 62 km/h


（上面第 4、5 条 print 各产生一个空行）
```

## 1.3 知识点 1.2　注释

### 1.3.1 核心概念

注释是写给人看的说明，**不参与运行**。它的价值在于让三个月后的自己、以及批改作业的教师，能在不读代码细节的情况下知道"这段在干什么"。

Python 有两种注释：

1. **单行注释**：以 `#` 开头，直到行尾都是注释；
2. **文档注释（三引号注释）**：用三个引号包住的一段多行文字，放在文件开头用于说明程序名称、作者、日期、功能。

### 1.3.2 语法要点

| 形式 | 写法 | 位置 | 用途 |
|---|---|---|---|
| 单行注释 | `# 输出标题` | 可独占一行，也可跟在语句后 | 解释某一行/某一段 |
| 行尾注释 | `print("a")  # 说明` | 语句右侧，`#` 前至少一个空格 | 说明该行输出含义 |
| 文档注释 | `"""……"""` | 文件第一行开始 | 程序头信息、函数说明 |

### 1.3.3 常见用法与易错点

- **注释要写"为什么"，不只写"是什么"**。`# 输出提示` 价值低；`# 分隔线统一 32 个等号，与大屏模板对齐` 才有信息量。
- 三引号注释**不是**真正的注释语句，它本质是一个未被赋值的字符串；放在文件或函数开头时，Python 会把它记录为该文件/函数的文档字符串（`__doc__`），第 10 章会用到这一特性。
- 注释不能嵌套：`# 外层 # 内层` 中第二个 `#` 仍属注释内容，不是新注释。
- 注释掉的代码不会执行，可用它做"临时停用某行"的调试手段。

### 1.3.4 随堂示例：注释不参与运行

**核心要点**：验证三类注释的实际效果：独立成行的注释、行尾注释、被注释掉的语句都不产生输出，只有未被注释的 `print` 才输出。

````python
# 这是独立成行的注释，Python 直接跳过
print("提示：前方施工，请减速")      # 行尾注释：说明本行含义
# print("这行被注释掉了，不会输出")
print("注释不影响程序结果")          # 三条 print 只生效两条
````

**预期输出**：

```text
提示：前方施工，请减速
注释不影响程序结果
```

## 1.4 知识点 1.3　转义字符与多行字符串

### 1.4.1 核心概念

有些字符无法直接写进字符串：换行、制表、引号本身、反斜杠。它们用**反斜杠加一个字母**表示，称为**转义字符**。Python 读到反斜杠时，不会原样输出，而是与后面的字符组合成一个特殊字符。

**多行字符串**用三个引号定义，字符串中的换行位置**就是**输出时的换行位置，因此可以用一条 `print` 输出整段文字。

### 1.4.2 语法要点

| 转义字符 | 含义 | 交通场景用途 | 示例输出 |
|---|---|---|---|
| `\n` | 换行 | 一条 `print` 内分行，如"自检完成 / 系统正常" | 光标移到下一行 |
| `\t` | 制表（跳到下一个制表位） | 车牌、时间、车位三列对齐 | 列间出现较大空隙 |
| `\\` | 一个反斜杠 | 输出 Windows 路径 `D:\traffic\...` | `D:\traffic` |
| `\"` | 一个双引号 | 文本中需要引号 | `"限速 60"` |
| `\'` | 一个单引号 | 外层用单引号时输出单引号 | `it's` |

**多行字符串写法**：

```text
print("""第一行
第二行
第三行""")
```

三条规则：① 第一个换行发生在 `"""` 之后；② 最后一个换行不产生，收尾 `"""` 要紧贴最后一行文字；③ 三引号内的缩进会原样输出。

### 1.4.3 常见用法与易错点

| 易错点 | 现象 | 原因与修正 |
|---|---|---|
| 路径只写一个反斜杠 | `D: raffic\parking`（`\t` 变成制表符） | `"D:\traffic"` 中的 `\t` 被解释为制表符，须写 `"D:\\traffic"` |
| 三引号内多余空行 | 输出首行或末行多一个空行 | 收尾 `"""` 紧贴文字；开头不留空行 |
| 三引号内缩进 | 每行前多出空格 | 三引号内容不要跟随代码缩进，或使用 `textwrap.dedent`（后续章节） |
| 以为 `\n` 会输出两个字符 | 实际只换行 | `\n` 是一个字符；要输出 `\n` 字样须写 `\\n` |

> **重要提示**：第 8 章读写文件路径、第 13 章绘图字体名都会再次遇到反斜杠与引号问题，本章务必把 `\\` 写对。

### 1.4.4 随堂示例：转义字符与多行字符串

**核心要点**：验证 `\n`（换行）、`\t`（制表）、`\\`（一个反斜杠）与三引号多行字符串的输出差异，重点是路径中的反斜杠必须写两个。

````python
# \n：在一条 print 内换行
print("自检完成\n系统正常")

# \t：制表位，让三列起点对齐
print("车牌\t时间\t车道")
print("京A12345\t08:15\t2")

# \\：输出一个反斜杠；若只写一个 \t，路径会被"吃掉"
print("日志路径：D:\\traffic\\log.csv")

# 三引号：源码换行即输出换行，收尾引号紧贴文字
print("""【入场须知】
一、限速 5 km/h
二、限高 2.1 m""")
````

**预期输出**：

```text
自检完成
系统正常
车牌	时间	车道
京A12345	08:15	2
日志路径：D:\traffic\log.csv
【入场须知】
一、限速 5 km/h
二、限高 2.1 m
```

## 1.5 知识点 1.4　输出排版

### 1.5.1 核心概念

排版的目标是"让人一眼看清结构"。本章只用手工排版，三种手段：

1. **分隔线**：用重复字符写成固定长度的字符串，如 32 个等号；
2. **居中**：在标题前后补空格。等宽字体下，中文字符占两列，标题左右空格数要按"（总宽 − 标题显示宽度）÷ 2"估算；
3. **对齐**：用 `\t` 让同一列的起始位置一致，或用空格手工补齐。

第 4 章会讲 `f-string` 的 `:<10` `:>10` 精确对齐，届时可替换本章的手工排版；本章先建立"列要对齐"的意识。

### 1.5.2 语法要点

| 手法 | 写法 | 说明 |
|---|---|---|
| 分隔线 | `print("=" * 32)` 属第 2 章；本章写 `print("================================")` | 直接数字符，保证上下两行等长 |
| 标题居中 | `print("        智慧交通出行提示")` | 前置 8 个空格，目视居中即可 |
| 列对齐 | `print("车牌号码\t入场时间\t车位编号")` | 表头与数据行使用**相同数量**的 `\t` |
| 收尾线 | `print("--------------------------------")` | 与标题线区分字符，形成视觉层次 |

### 1.5.3 常见用法与易错点

- **上下分隔线必须等长**，否则版式歪斜；用编辑器列号或数一数字符确认。
- 表头与数据行的 `\t` 数量必须一致；若某列文字宽度差异过大，可改用两个 `\t`，但要**整列统一**改。
- 中英文混排时，汉字占两列而英文字母占一列，`\t` 的对齐效果会出现轻微参差，属正常现象；正式报表应在第 4 章用格式化宽度控制，或改用表格控件。

### 1.5.4 随堂示例：分隔线与两列对齐

**核心要点**：验证手工排版的三条规则：上下分隔线字符数相同、标题两侧补空格居中、表头与数据行的 `\t` 数量一致。

````python
# 上下分隔线用相同数量的等号，版式才不歪
print("==============================")
print("        智慧交通出行提示")
print("==============================")

# 表头与数据行都用一个 	 分隔两列
print("车牌号码	车位编号")
print("京A12345	B-013")
print("京B67890	A-042")

# 换一种字符做次级分隔，形成视觉层次
print("------------------------------")
print("发布单位：市交通运行监测中心")
````

**预期输出**：

```text
==============================
        智慧交通出行提示
==============================
车牌号码	车位编号
京A12345	B-013
京B67890	A-042
------------------------------
发布单位：市交通运行监测中心
```

## 1.6 知识点 1.5　程序的保存与运行

### 1.6.1 核心概念

一段代码要能被反复运行，必须**存成文件**。Python 源文件以 `.py` 为扩展名，文件内容就是纯文本程序。本章提供两种运行方式：

1. **Jupyter 单元格**：改一格、按 `Shift + Enter` 立刻看结果，适合课堂跟做；
2. **脚本文件**：在终端执行 `python 文件名.py`，适合提交作业与真实部署。

### 1.6.2 语法要点

| 事项 | 规范 | 说明 |
|---|---|---|
| 文件扩展名 | `.py` | 不可写成 `.txt` |
| 文件编码 | UTF-8 | 含中文的源文件必须用 UTF-8 保存，否则运行时报 `SyntaxError: Non-UTF-8 code` |
| 文件命名 | 英文或拼音 + 下划线，如 `case1_信息板提示.py` | 避免空格与特殊符号，便于命令行执行 |
| 运行命令 | `python 文件名.py` | 需在文件所在目录执行，或用绝对路径 |
| 输出位置 | 终端 / 单元格下方 | 脚本运行结果显示在终端，不是弹窗 |

### 1.6.3 常见用法与易错点

- 双击 `.py` 文件可能一闪而过：程序运行完窗口立即关闭。应在终端中运行，或在末尾加 `input()` 暂停（第 2 章）。
- `python: command not found`：命令名可能是 `python3`，用 `python3 文件名.py` 试。
- Jupyter 中修改代码后未重新运行，看到的是上一次结果；判断依据是单元格左侧的执行序号 `[n]` 是否递增。
- 本章代码无输入，可放心"运行全章"。

## 答案 1-2　车载仪表信息牌

> 对应 **练习 1-2**（第三部分）

### 参考代码

````python
# 车载仪表自检信息输出
print("项目\t数值\t单位")          # 表头：三列名称
print("车速\t62\tkm/h")            # 实时车速
print("剩余电量\t78\t%")            # 动力电池 SOC
print("续航\t316\tkm")              # 剩余续航里程
print("挡位\tD\t-")                # 当前挡位

# 一条 print 内用 \n 输出两行收尾信息
print("自检完成\n系统正常，可以起步")
````

### 运行结果

```text
项目	数值	单位
车速	62	km/h
剩余电量	78	%
续航	316	km
挡位	D	-
自检完成
系统正常，可以起步
```

### 解析

1. **对齐的实现**：5 行都使用**两个** `\t`（三列之间各一个），列起点因此一致。若某行多写或少写一个 `\t`，该行会整体右移或左移。
2. **`\t` 与空格的取舍**：`\t` 跳到下一个制表位，列宽随内容自动调整，比手工数空格稳健；但"剩余电量"四字比"车速"两字宽，第二列仍会略向右，属中英文混排的正常表现（见知识点 1.4 的说明）。
3. **`\n` 的用法**：`"自检完成\n系统正常，可以起步"` 是**一个**字符串，`\n` 在输出时才换行，因此只调用一次 `print` 就得到两行——这正是题目第 3 条要求的用意：区分"两条语句"与"一个字符串内的换行"。
4. **行尾注释**：注释与代码之间留两个空格，便于阅读；注释说明"这一行代表什么业务含义"，而不是重复代码内容。

### 常见错误

| 错误现象 | 原因 | 修正 |
|---|---|---|
| 列不对齐 | 各行 `\t` 数量不一致 | 统一为每行两个 `\t` |
| 收尾信息只输出一行 | 写成 `print("自检完成" "系统正常")` 相邻字符串自动拼接 | 中间用 `\n` |
| `%` 导致报错 | 误用 `%` 格式化语法 | 本章 `%` 只是普通字符，直接写在字符串内 |
| 输出 `km/h` 变成两行 | 误写成 `\n` 或斜杠后接 `n` | 检查是否误敲 `\n` |

| 项 | 分值 | 观察点 |
|---|---|---|
| 三列对齐 | 4 | 表头与 4 行数据列起点一致 |
