# 第一部分　知识讲解

## 5.1 场景导入：四类容器，四种需求

一条车道上是"一串车"，一批设备是"一组编号"，每辆车的档案是"键—值"，重复上报的记录需要"去重"——这四种需求恰好对应四类容器：

| 容器 | 有序 | 可变 | 允许重复 | 访问方式 | 交通场景 |
|---|---|---|---|---|---|
| 列表 `list` | 是 | 是 | 是 | 下标 | 车辆队列、时段序列 |
| 元组 `tuple` | 是 | **否** | 是 | 下标 | 坐标、配时方案等固定参数 |
| 集合 `set` | 否 | 是 | **否** | 只能遍历 / 判断包含 | 去重、交集比对 |
| 字典 `dict` | 否（3.7+ 保插入序） | 是 | 键不重复 | **键** | 台账、配置表、统计表 |

选择方法（四问）：**要不要顺序？会不会修改？要不要去重？按位置还是按键取？**

## 5.2 知识点 5.1　列表的创建与增删改查

### 5.2.1 核心概念

列表用方括号定义，元素之间用逗号分隔，元素类型可以不同（实践中建议同类）。列表是**可变**的：可以增、删、改，且改动**原地生效**，不需要重新赋值。

### 5.2.2 语法要点

| 操作 | 写法 | 说明 | 易错点 |
|---|---|---|---|
| 创建 | `queue = ["京A12345", "京B67890"]` | 也可 `list()` 转换 | —— |
| 索引 | `queue[0]`、`queue[-1]` | 与字符串一致 | 越界抛 `IndexError` |
| 切片 | `queue[:2]`、`queue[1:]` | 返回**新列表** | 原列表不变 |
| 尾部追加 | `queue.append(x)` | 一次只能加**一个**元素 | `append([a, b])` 会把整个列表当一个元素 |
| 指定插入 | `queue.insert(1, x)` | 原位置及之后元素后移 | 下标越界不报错，插到末尾 |
| 按下标删除 | `queue.pop(0)` | 取出并删除；不写下标默认删**最后一个** | 想删队首必须写 `pop(0)` |
| 按值删除 | `queue.remove(x)` | 删除第一个匹配值 | 值不存在抛 `ValueError` |
| 长度 | `len(queue)` | 元素个数 | —— |
| 修改 | `queue[0] = "京X00000"` | 按下标赋值 | 只能改已存在的下标 |

### 5.2.3 常见用法与易错点

- **`append` 与 `+` 的区别**：`append` 原地加一个元素；`lst1 + lst2` 返回新列表。
- **`pop()` 的默认行为**：不写下标时删除**最后一个**，这与"队列"的直觉相反，队首出队必须 `pop(0)`。
- **`remove()` 只删第一个匹配**：列表中有重复值时需循环删除，或改用推导式过滤。
- **遍历时删除元素**会导致漏项：边遍历边 `remove` 会跳过元素，正确做法是新建列表或倒序遍历。
- 列表可以嵌套（第 6 章的路网树、第 13 章的车位矩阵都用到）。

### 5.2.4 随堂示例：队列的入队、插队与放行

**核心要点**：验证列表四个高频方法的分工：`append()` 尾部追加、`insert(i, x)` 指定位置插入、`pop()` 取出并删除、`remove(x)` 按值删除。

````python
queue = ["京A12345", "京B67890"]
print("初始队列：", queue)

queue.append("津C24680")          # 队尾加入
queue.insert(0, "冀D11223")       # 插到队首（应急车辆优先）
print("插入后：", queue, "| 共", len(queue), "辆")

left = queue.pop(0)               # 放行队首
print("放行：", left, "| 剩余：", queue)

queue.remove("京B67890")          # 按车牌剔除异常车辆
print("剔除后：", queue)
````

**预期输出**：

```text
初始队列： ['京A12345', '京B67890']
插入后： ['冀D11223', '京A12345', '京B67890', '津C24680'] | 共 4 辆
放行： 冀D11223 | 剩余： ['京A12345', '京B67890', '津C24680']
剔除后： ['京A12345', '津C24680']
```

## 5.3 知识点 5.2　排序与统计

### 5.3.1 核心概念

| 函数/方法 | 作用 | 是否改原列表 | 返回值 |
|---|---|---|---|
| `lst.sort()` | 原地升序 | **是** | `None` |
| `lst.sort(reverse=True)` | 原地降序 | 是 | `None` |
| `sorted(lst)` | 返回排序后的**新列表** | 否 | 新列表 |
| `len(lst)` | 元素个数 | 否 | int |
| `sum(lst)` | 求和 | 否 | 数值 |
| `max(lst)` / `min(lst)` | 最大 / 最小 | 否 | 元素 |
| `lst.index(x)` | 首次出现的下标 | 否 | int（不存在抛 `ValueError`） |
| `lst.count(x)` | 出现次数 | 否 | int |

### 5.3.2 语法要点

- `sort()` 返回 `None`，写成 `x = lst.sort()` 会让 `x` 变成 `None`——这是最常见的错误之一。
- `sorted()` 可用于任何可迭代对象，且能配合 `key=`（第 7 章重点）。
- `max` / `min` 配合 `index` 可求"峰值及其位置"：`flows.index(max(flows))`。
- 统计平均值：`sum(lst) / len(lst)`，注意 `len` 为 0 时会除零。

### 5.3.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| `x = lst.sort()` | `x` 为 `None` | 直接 `lst.sort()`，或改用 `x = sorted(lst)` |
| 想保留原顺序却用了 `sort()` | 原列表被打乱 | 用 `sorted()` |
| `max([])` | `ValueError: max() arg is an empty sequence` | 先判空 |
| `index` 找不存在的值 | `ValueError` | 先 `in` 判断 |
| 混排数字与字符串 | `TypeError: '<' not supported` | 统一类型后再排序 |

### 5.3.4 随堂示例：sort() 与 sorted() 的区别

**核心要点**：验证 `sort()` 原地排序且返回 `None`、`sorted()` 返回新列表不动原列表；同时用 `sum` / `max` / `min` 做统计。

````python
flows = [820, 1100, 760]

print("总量：", sum(flows), "| 最大：", max(flows), "| 最小：", min(flows))
print("峰值下标：", flows.index(max(flows)))

flows.sort()                      # 原地升序，返回 None
print("sort 后原列表：", flows)
print("sort 的返回值：", flows.sort())

ranked = sorted(flows, reverse=True)   # 返回新列表
print("sorted 降序：", ranked)
print("原列表仍是：", flows)
````

**预期输出**：

```text
总量： 2680 | 最大： 1100 | 最小： 760
峰值下标： 1
sort 后原列表： [760, 820, 1100]
sort 的返回值： None
sorted 降序： [1100, 820, 760]
原列表仍是： [760, 820, 1100]
```

## 5.4 知识点 5.3　元组

### 5.4.1 核心概念

元组用圆括号定义，**创建后不能修改**。它的价值不在"少写两个字符"，而在**表达"这组值不该被改动"**：路口坐标、信号配时方案、RGB 颜色，都属于"一经设定就不该被程序随手改掉"的数据。

### 5.4.2 语法要点

| 操作 | 写法 | 说明 |
|---|---|---|
| 创建 | `cross = ("京港澳 K12", 39.9042, 116.4074)` | 括号可省略，但建议保留 |
| 单元素元组 | `t = (5,)` | **必须带逗号**，`(5)` 只是整数 5 |
| 索引 / 切片 | `cross[0]`、`cross[:2]` | 与列表一致 |
| 解包 | `name, lat, lng = cross` | 元素个数必须与变量数一致 |
| 长度 | `len(cross)` | —— |
| 修改 | `cross[0] = "x"` | ❌ `TypeError: 'tuple' object does not support item assignment` |

### 5.4.3 常见用法与易错点

- **单元素元组漏逗号**：`(5)` 是 int，`(5,)` 才是元组；`type((5))` 返回 `<class 'int'>`。
- 元组可作**字典的键**（列表不行），因为键必须不可变——这是第 7 章"按元组多维排序"的基础。
- 元组内若含列表，列表本身仍可修改（浅层不可变），不要依赖它做深度保护。
- 函数返回多个值时，实际返回的就是元组（第 6 章）。

### 5.4.4 随堂示例：元组：不可变的固定参数

**核心要点**：验证元组用下标访问与切片读取，但**不能修改元素**；单元素元组必须写逗号，否则只是加了括号的普通值。

````python
cross = ("京港澳 K12", 39.9042, 116.4074)
print("点位：", cross[0])
print("纬度：", cross[1], "| 经度：", cross[2])
print("字段数：", len(cross))

# 元组不可改：下面这行一旦取消注释就会报 TypeError
# cross[0] = "京开 K8"

single = (62,)            # 有逗号才是元组
not_tuple = (62)          # 没有逗号只是整数
print("单元素元组：", single, type(single))
print("加括号的整数：", not_tuple, type(not_tuple))
````

**预期输出**：

```text
点位： 京港澳 K12
纬度： 39.9042 | 经度： 116.4074
字段数： 3
单元素元组： (62,) <class 'tuple'>
加括号的整数： 62 <class 'int'>
```

## 5.5 知识点 5.4　集合

### 5.5.1 核心概念

集合用花括号定义，特点是**无序**、**元素不重复**。两大用途：

1. **去重**：`set(records)` 一步得到唯一值；
2. **集合运算**：交集 `&`、并集 `|`、差集 `-`、对称差 `^`。

### 5.5.2 语法要点

| 操作 | 写法 | 示例结果（A={"京A","京B"}，B={"京B","津C"}） |
|---|---|---|
| 创建 | `s = {"京A12345", "京B67890"}` | —— |
| 空集合 | `set()` | **不能写 `{}`**（那是空字典） |
| 去重 | `set(lst)` | 元素唯一 |
| 交集 | `A & B` | `{"京B"}` |
| 并集 | `A \| B` | `{"京A","京B","津C"}` |
| 差集 | `A - B` | `{"京A"}` |
| 增删 | `s.add(x)`、`s.discard(x)` | `discard` 不报错，`remove` 不存在时报错 |

### 5.5.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| `{}` 当空集合 | 得到空字典 | 用 `set()` |
| 期望集合有序 | 输出顺序与插入顺序无关，且每次运行可能不同 | 需稳定输出用 `sorted(s)` |
| 用集合存可重复数据 | 重复项被吞 | 需要重复用列表 |
| 用 `s[0]` 取元素 | `TypeError: 'set' object is not subscriptable` | 集合不支持下标，只能遍历或 `list(s)[0]` |

> **工程提示**：集合的遍历顺序受哈希随机化影响，输出不稳定。凡是要打印给用户看或用于批改，一律 `sorted(set(...))`，案例 5-2 与练习 5-2 都遵循此约定。

### 5.5.4 随堂示例：集合去重与交并差

**核心要点**：验证集合自动去重（无序、不重复），以及交集 `&`、并集 `|`、差集 `-` 的语义——这是比对两个时段过车记录的标准写法。

````python
this_hour = ["京A12345", "京B67890", "京A12345"]
a = set(this_hour)
print("原始条数：", len(this_hour), "→ 去重后：", len(a))

b = {"京B67890", "津C24680"}
print("本时段车辆：", sorted(a))
print("上时段车辆：", sorted(b))
print("两时段都出现（交集）：", sorted(a & b))
print("任一时段出现（并集）：", sorted(a | b))
print("仅本时段新增（差集）：", sorted(a - b))
````

**预期输出**：

```text
原始条数： 3 → 去重后： 2
本时段车辆： ['京A12345', '京B67890']
上时段车辆： ['京B67890', '津C24680']
两时段都出现（交集）： ['京B67890']
任一时段出现（并集）： ['京A12345', '京B67890', '津C24680']
仅本时段新增（差集）： ['京A12345']
```

## 5.6 知识点 5.5　字典

### 5.6.1 核心概念

字典保存"键 → 值"的映射，用键快速查找值，时间复杂度接近 O(1)。它是交通台账、配置表、统计汇总的通用结构。

### 5.6.2 语法要点

| 操作 | 写法 | 说明 |
|---|---|---|
| 创建 | `ledger = {"京A12345": ("小型车", 62)}` | 键必须不可变（str / int / tuple） |
| 取值 | `ledger["京A12345"]` | 键不存在抛 `KeyError` |
| 安全取值 | `ledger.get("京X", 默认值)` | 不存在返回默认值，**推荐对外使用** |
| 新增 / 修改 | `ledger["冀D11223"] = ("新能源", 55)` | 键存在即修改，不存在即新增 |
| 删除 | `del ledger["津C24680"]` / `pop(键)` | `pop` 可带默认值避免报错 |
| 判断键 | `"津C24680" in ledger` | 判断的是**键**，不是值 |
| 键列表 | `ledger.keys()` | —— |
| 值列表 | `ledger.values()` | —— |
| 键值对 | `ledger.items()` | 用于 `for k, v in d.items()` |
| 长度 | `len(ledger)` | 键值对个数 |

### 5.6.3 常见用法与易错点

| 易错点 | 现象 | 修正 |
|---|---|---|
| 用不存在的键取值 | `KeyError` | 用 `get(键, 默认值)` |
| `"值" in d` 想判断值 | 实际判断的是键 | 判断值用 `值 in d.values()` |
| 边遍历边增删 | `RuntimeError: dictionary changed size during iteration` | 先取 `list(d.keys())` 再遍历（第 6 章） |
| 字典累加未设初值 | `KeyError` | `d[k] = d.get(k, 0) + v`（第 8 章练习） |
| 排序字典 | 字典本身无序 | 对 `items()` 用 `sorted()`（第 7 章） |

### 5.6.4 随堂示例：字典：按路段查路况

**核心要点**：验证字典「键—值」的增、查、默认值 `get()` 与遍历 `items()`；重点体会 `[]` 取不到键会报错，而 `get()` 可以给默认值。

````python
status = {"京开": "缓行", "机场": "畅通"}

status["京港澳"] = "拥堵"            # 新增键值对
print("路段数：", len(status))
print("京开的状态：", status["京开"])

# get()：键不存在时返回默认值，不报错
print("京承（get）：", status.get("京承", "无数据"))

print("逐条遍历：")
for road, level in status.items():
    print("  ", road, "→", level)
````

**预期输出**：

```text
路段数： 3
京开的状态： 缓行
京承（get）： 无数据
逐条遍历：
   京开 → 缓行
   机场 → 畅通
   京港澳 → 拥堵
```

## 5.7 知识点 5.6　列表推导式

### 5.7.1 核心概念

推导式用一行代码"由旧列表生成新列表"，形式为：

```text
[表达式 for 变量 in 序列 if 条件]
```

它等价于"建空列表 → 循环 → 判断 → 追加"，但更紧凑。第 7 章的 `map` / `filter` 与它功能重叠，届时可对比选择。

### 5.7.2 语法要点

| 写法 | 含义 | 示例结果（flows 见案例） |
|---|---|---|
| `[x for x in flows]` | 复制 | 原样 |
| `[x for x in flows if x > 1500]` | 筛选 | `[1650]` |
| `[x // 100 for x in flows]` | 变换 | 各值折百 |
| `[x * 2 for x in flows if x > 700]` | 先筛后变 | 大于 700 的值加倍 |

### 5.7.3 常见用法与易错点

- 推导式只能写**表达式**，不能写语句（不能赋值、不能 `print`）。
- 条件放在**后面**（`if` 在 `for` 之后），不要写成 `if` 在前。
- 超过两层的嵌套推导式应改回普通循环，可读性优先。
- 推导式产生**新列表**，原列表不变。

---

### 5.7.4 随堂示例：列表推导式一行完成筛选与换算

**核心要点**：验证推导式 `[表达式 for 变量 in 序列 if 条件]` 的三个位置各自的作用：表达式决定元素形态、`if` 决定保留哪些元素。

````python
flows = [820, 1100, 760, 1650]

over = [f for f in flows if f > 1000]
print("超 1000 的时段：", over, "| 共", len(over), "个")

half = [f // 100 for f in flows]
print("折合百辆：", half)

labels = ["高" if f > 1500 else "平" for f in flows]
print("等级标记：", labels)
````

**预期输出**：

超 1000 的时段： [1100, 1650] | 共 2 个

````python
# ---------- 第 05 章 列表、元组、集合与字典 · 案例 5-1：进口道车辆队列管理 ----------
# 对应知识点：5.1 列表增删改查（append / insert / pop / remove）、5.2 sort() 与统计、5.6 列表推导式
# 功能说明：模拟进口道队列的入队、应急车辆插队、绿灯放行与异常剔除，并对车速排序统计

# 1) 建立队列并查看状态
queue = ["京A12345", "京B67890", "津C24680", "京A13579"]
print("初始队列：", queue, "共", len(queue), "辆")
print("队首：", queue[0], "| 队尾：", queue[-1])
print("前两辆：", queue[:2])

# 2) 入队与插队
queue.append("冀D11223")          # 队尾加入
print("新车入队后：", queue)
queue.insert(1, "京E55667")       # 插到第 2 位（应急车辆优先）
print("应急车辆插入后：", queue)

# 3) 放行与剔除
left = queue.pop(0)               # 取出并删除队首
print("绿灯放行：", left)
queue.remove("津C24680")          # 按车牌剔除异常车辆
print("剔除后队列：", queue, "共", len(queue), "辆")

# 4) 车速排序与统计
speeds = [62, 45, 78, 51, 66]
speeds.sort()                     # 原地升序
print("车速升序：", speeds)
print("最高：", max(speeds), "最低：", min(speeds))
print("平均：", round(sum(speeds) / len(speeds), 1))

speeds.sort(reverse=True)         # 原地降序
print("车速降序：", speeds)
print("前三快：", speeds[:3])
````

# 第四部分　习题答案

## 答案 5-1　分时段车流量统计分析

> 对应 **练习 5-1**（第三部分）

### 参考代码

````python
# ---------- 第 05 章 列表、元组、集合与字典 · 练习 5-1 参考答案：分时段车流量统计 ----------
# 对应知识点：5.2 sum / max / min / index 统计、5.1 切片、5.6 列表推导式
# 功能说明：统计 8 个时段车流量的总量、均值、峰谷位置与早晚高峰合计，并折算百辆

flows = [820, 1100, 760, 690, 720, 980, 1650, 1320]

print("时段数：", len(flows))
print("总流量：", sum(flows), "辆")
print("平均：", round(sum(flows) / len(flows), 1), "辆")
print("最大：", max(flows), "出现在第", flows.index(max(flows)) + 1, "个时段")
print("最小：", min(flows), "出现在第", flows.index(min(flows)) + 1, "个时段")

morning = sum(flows[:4])          # 切片取前 4 个时段：早高峰
evening = sum(flows[4:])          # 后 4 个时段：晚高峰
print("早高峰合计：", morning, "| 晚高峰合计：", evening)
print("晚高峰是否更高：", evening > morning)

over = [x for x in flows if x > 1500]
print("超 1500 的时段流量：", over, "共", len(over), "个")

hundreds = [x // 100 for x in flows]
print("折合百辆：", hundreds)
````

### 运行结果

```text
时段数： 8
总流量： 8040 辆
平均： 1005.0 辆
最大： 1650 出现在第 7 个时段
最小： 690 出现在第 4 个时段
早高峰合计： 3370 | 晚高峰合计： 4670
晚高峰是否更高： True
超 1500 的时段流量： [1650] 共 1 个
折合百辆： [8, 11, 7, 6, 7, 9, 16, 13]
```

### 解析

1. **`index` 与业务编号差 1**：`flows.index(max(flows))` 返回 6，而业务上说"第 7 个时段"，因此输出时 `+1`。这是"程序下标从 0 起、业务编号从 1 起"的典型转换。
2. **切片求分段合计**：`flows[:4]` 取下标 0—3（前 4 个），`flows[4:]` 取下标 4 之后（后 4 个），两段互补、不重叠。
3. **推导式的两种用法**：`[x for x in flows if x > 1500]` 是"筛选"，`[x // 100 for x in flows]` 是"变换"。筛选条件写在 `for` 之后，不要前置。
4. **为什么用 `//` 折百**：÷100 后取整，`820 // 100` 得 8，属"百辆"口径；若用 `round(820/100)` 得 8 但 `1650` 会得 17（四舍五入），与"折合整百"的业务口径不符。

### 常见错误

| 错误现象 | 原因 | 修正 |
|---|---|---|
| 峰值显示"第 6 个时段" | 忘记 `+1` | 业务编号 `+1` |
| 早高峰合计含第 5 个时段 | 切片写成 `[:5]` | 前 4 个写 `[:4]` |
| 超阈值列表含 1320 | 条件写成 `>= 1500` 或阈值写错 | `> 1500` |
| `TypeError` | 推导式里写成 `if` 在前 | `[x for x in flows if ...]` |

## 答案 5-2　路段拥堵状态查询与汇总

> 对应 **练习 5-2**（第三部分）

### 参考代码

````python
# ---------- 第 05 章 列表、元组、集合与字典 · 练习 5-2 参考答案：路段拥堵状态查询与汇总 ----------
# 对应知识点：5.5 字典遍历 / get() / values()、5.1 列表收集、5.4 集合去重
# 功能说明：按路段查询路况（含缺失路段的默认值），汇总拥堵路段并统计出现过的状态种类

status = {
    "京港澳高速": "拥堵",
    "京开高速": "缓行",
    "机场高速": "畅通",
}

for road in status:
    print(road, "->", status[road])

print("京开高速状态：", status["京开高速"])
print("机场高速状态：", status.get("机场高速", "无数据"))
print("京承高速状态：", status.get("京承高速", "无数据"))

crowded = []
for road in status:
    if status[road] == "拥堵":
        crowded.append(road)
print("拥堵路段：", crowded, "共", len(crowded), "条")

print("全部畅通：", "拥堵" not in status.values() and "缓行" not in status.values())  # values() 取出所有状态值
print("状态种类：", sorted(set(status.values())))
````

### 运行结果

```text
京港澳高速 -> 拥堵
京开高速 -> 缓行
机场高速 -> 畅通
京开高速状态： 缓行
机场高速状态： 畅通
京承高速状态： 无数据
拥堵路段： ['京港澳高速'] 共 1 条
全部畅通： False
状态种类： ['拥堵', '畅通', '缓行']
```

### 解析

1. **遍历字典得到键**：`for road in status` 每次得到键，再用 `status[road]` 取值；也可写 `for road, st in status.items()`（第 7 章的 `enumerate` / `items` 技巧）。
2. **`get` 的缺失保护**：`get("京承高速", "无数据")` 不报错且给出可读结论，是面向用户接口的必备写法；`[]` 只适合"确定键存在"的内部场景。
3. **收集模式**：空列表建在循环外，循环内 `append`，循环后统一输出——与第 3 章"累加器"的思路一致，只是把加法换成追加。
4. **`values()` 判断**：`"拥堵" not in status.values()` 直接对值集合做包含判断，比遍历计数更简洁；`and` 连接两个条件，任一不满足即为 `False`。
5. **`sorted(set(...))` 去重排序**：状态只有三种，用集合去重、排序后输出，顺序稳定。

### 常见错误

| 错误现象 | 原因 | 修正 |
|---|---|---|
| `KeyError` | 用 `status["京承高速"]` | 改 `get(键, 默认值)` |
| 拥堵列表为空 | 遍历时用了 `road` 而非 `status[road]` 比较 | 比较值要取 `status[road]` |
| 列表只保留最后一条 | `crowded = []` 写在循环内 | 移到循环外 |
| 状态种类顺序乱 | 直接打印 `set` | `sorted(set(...))` |

## 本章小结

- **选容器四问**：要顺序？会修改？要去重？按位置还是按键？
- **列表**：`append` 加尾、`insert` 插位、`pop(0)` 出队首、`remove` 按值删；`sort()` 原地、`sorted()` 返回新表。

- **字典**：键值映射，`get()` 做缺失保护，`values()` / `items()` 用于遍历；`in` 判断的是键。
